## 1. NumPy arrays


**NumPy** stores numerical values in arrays and carries out calculations on them.
A one-dimensional array could hold voltage readings over time; a two-dimensional
array could hold an image or a table with one row per observation.

For the first examples, pay attention to three things:

- **Shape:** how many values there are along each dimension, or **axis**.
- **Data type (`dtype`):** how values are stored, for example as integers or
  floating-point numbers. This affects precision and the range of values possible.
- **Indexing:** which rows, columns or individual values an expression selects.

An expression such as `2 * readings` multiplies every value by two. This is an
**element-wise operation**. Using array expressions instead of writing a Python
loop is often called **vectorisation**. It can reduce calculation time, but the
gain depends on the task; Section 3 shows how to measure it.

**Broadcasting** lets compatible arrays of different shapes be combined, for
example subtracting one background value from every detector reading. The
examples below show the shapes involved. They also cover statistics and the
matrix operations used in physics calculations.


### 1.1 Creating arrays


```python
# Create an array from a list of rows. The shape is (2, 3): two rows and three columns.
A = np.array([[1,2,3],[4,5,6]])
print("This is array A:\n",A)
```

```python
# zeros and ones create arrays with known starting values; dtype chooses the number type.
# np.empty reserves space without setting its values. Fill it before calculating,
# or use zeros when you need a known starting value.
A = np.zeros((2,3))

B = np.ones(5) 

C = np.zeros(shape=(100,100), dtype=float)

D = np.ones((2,2), dtype=np.int16)

E = np.empty((2,2))

print("Array of zeros A:\n", A)
print("\nArray of ones B:", B)
print("\nLarge array C shape:", C.shape, "with dtype:", C.dtype)
print("\nInteger array D:\n", D)
print("\nEmpty array E (uninitialised values; not random samples):\n", E)
```

```python
# ===== Creating Arrays from Ranges =====
# Similar to range() in Python, but creates NumPy arrays

# arange(start, stop, step) - like range() but for arrays
# Creates [2, 4, 6, 8] - stops BEFORE 10
A = np.arange(2, 10, 2)

# linspace(start, stop, num_points) - linearly spaced points
# Creates 10 evenly-spaced values from 0 to 4.5 (INCLUDES endpoint)
# Useful for creating x-axes for plotting
B = np.linspace(0, 4.5, 10)

print("arange result A:", A)
print("linspace result B:", B)
```

```python
# rand gives uniform values from 0 up to, but not including, 1.
# randn gives a normal distribution with mean 0 and standard deviation 1.
# Multiplying by 2 and adding 10 changes these to standard deviation 2 and mean 10.
A = np.random.rand(1, 2)  # 1 row, 2 columns of uniform random [0, 1)

B = 2 * np.random.randn(1, 2) + 10  # mean=10, std=2

print("Uniform random [0,1):", A)
print("Normal random (μ=10, σ=2):", B)
```

```python
# Inspect dimensions, shape and number type before calculating.
# size counts all entries; itemsize gives bytes per entry.
# .data describes the underlying storage and is rarely needed for analysis.
print("Number of dimensions (ndim):", A.ndim)      # e.g., 1D vector, 2D matrix, 3D tensor
print("Shape (rows, cols, ...):", A.shape)         # length along each dimension
print("Total elements (size):", A.size)            # total number of elements
print("Data type (dtype):", A.dtype)               # int64, float64, etc.
print("Bytes per element (itemsize):", A.itemsize) # memory usage per number
print("Memory buffer (data):", A.data)             # description of underlying storage
```

### 1.2 Reshaping, slicing and combining arrays

```python
# Slice positions 2, 3, 4: the start is included and the stop is excluded.
A = np.arange(10)
A[2:5]
```

```python
# Looping over a 2D array gives one row at a time. Square each value in that row.
for i in A: 
    print (i**2)
```

```python
# The list comprehension collects one squared row for each row of A.
[i**2 for i in A]
```

```python
# A.flat supplies individual values across the array instead of whole rows.
[i**2 for i in A.flat]
```

```python
# ravel lays the values out in one dimension; reshape groups them into a new shape.
# The number of entries stays the same. .T exchanges rows and columns of this 2D array.
A = np.floor(10*np.random.random((3,4)))
print(A, A.shape)
print(A.ravel()) # flatten array
print(A.reshape(6,2)) 
print(A.T, A.T.shape)
```

```python
# B and A name the same array. resize changes that array's shape directly.
# reshape returns an array with the requested shape; here its result is not stored.
# In reshape(3, -1), NumPy calculates the number of columns needed for three rows.
B = A 
A.resize(2,6)
print(A)
B.reshape(3,-1) # -1 calc cols automatically
print(B)
```

```python
# vstack places B's rows below A; hstack places B's columns beside A.
# The dimensions along the unchanged direction must match.
A = np.floor(10*np.random.random((3,4)))
B = np.floor(10*np.random.random((3,4)))
print(A)
print(B)
C = np.vstack((A,B))
print(C)
D = np.hstack((A,B))
print(D)
```

```python
# hsplit divides columns into groups. A number requests equal-sized groups;
# a sequence specifies positions at which new groups begin.
A = np.floor(10*np.random.random((2,12)))
print("A", A)
B, C, D = np.hsplit(A,3)
print("B", B)
print("C", C)
print("D", D)
E, F, G = np.hsplit(A,(3,4)) # split before positions 3 and 4
print("E", E)
print("F", F)
print("G", G)
```

### 1.5 Array operations and broadcasting


```python
# Add corresponding entries: C[i, j] = A[i, j] + B[i, j].
B = A
C = A + B
print("Element-wise addition:\n", C)
```

```python
# Multiply every entry by 4. For arrays, * works entry by entry; @ means matrix multiplication.
C = A * 4
print("Multiply every element by 4:\n", C)
```

```python
# This polynomial uses arithmetic that NumPy can apply to an entire array.
# f(A) therefore evaluates the polynomial at every value in A.
def f(x):
    return 3*x**2 - 2*x + 7

f(A)
```

```python
# A *= 2 edits the existing array; other names or views sharing its values see the change.
# A = A * 2 would instead assign a newly calculated array.
A *= 2  # Multiply all elements by 2 and store result in A (no copy made)
print("After in-place multiplication by 2:\n", A)
```

## 2. Advanced NumPy


This section extends the array operations in Section 1. Use it when you need
to select measurements using several conditions, understand why changing one
array also changes another, or calculate with several array dimensions.

A **view** uses the same stored values as another array; a **copy** has its own
values. This distinction matters when keeping an original set of measurements
while making a corrected version. **Structured arrays** store named entries
with different types, such as an event number alongside a measured energy.

The later examples show how arrays are arranged in memory and introduce
`einsum`, which expresses sums using labelled indices. You can skip these on a
first reading. When using them, write down what each axis represents and check
the result against a small calculation you can verify by hand.


### 2.1 Select entries using index arrays and True/False conditions


```python
# Arrays of row and column positions select entries in pairs.
# A Boolean mask selects entries satisfying a condition. Use & for both conditions
# and put parentheses around each comparison.
arr = np.arange(20).reshape(4, 5)
print("Original array:")
print(arr)

row_indices = np.array([0, 2, 3])
col_indices = np.array([1, 3, 4])
print(f"\nFancy indexing: arr[{row_indices}, {col_indices}]")
# Pair the positions: read (row 0, column 1), then (2, 3), then (3, 4). This selects three entries,
# rather than every combination of these rows and columns.
print(arr[row_indices, col_indices])

mask = (arr > 5) & (arr < 15)
print(f"\nValues between 5 and 15:")
# The mask has the same shape as arr; the selected values form a one-dimensional array.
print(arr[mask])
```

### 2.2 Views and copies: when arrays share their values


```python
# Advanced slicing and views
arr = np.arange(24).reshape(4, 6)
print("Original array:")
print(arr)

# Slice creates a view (not a copy)
# Rows 1 and 2, columns 2, 3 and 4: the stop positions are excluded.
slice_view = arr[1:3, 2:5]
print(f"\nSlice (view):")
print(slice_view)

# Modifying view modifies original
# Position [0, 0] within this smaller view is position [1, 2] in arr.
slice_view[0, 0] = 999
print(f"\nOriginal array after modifying view:")
print(arr)

# Use .copy() to create independent array
arr_copy = arr[1:3, 2:5].copy()
arr_copy[0, 0] = -1
print(f"\nCopy is independent (original unchanged):")
print(arr)
```

### 2.3 Structured arrays: named entries with different data types


```python
# A structured array gives each record named entries of different types.
# U10 means text up to ten characters, i4 a 32-bit integer and f8 a 64-bit float.
# Select an entry across all records using its name, such as people['age'].
dt = np.dtype([('name', 'U10'), ('age', 'i4'), ('weight', 'f8')])

people = np.array([
    ('Alice', 25, 55.5),
    ('Bob', 30, 75.2),
    ('Charlie', 35, 80.1)
], dtype=dt)

print("Structured array (like a simple database):")
print(people)

print(f"\nAll names: {people['name']}")
print(f"Average age: {people['age'].mean()}")
print(f"Total weight: {people['weight'].sum()}")
```

### 2.4 Broadcasting: combine compatible array shapes


```python
# Broadcasting matches dimensions from the right: lengths must match or one must be 1.
# A length-one dimension is reused. Inspect the shapes before each addition.
matrix = np.ones((4, 3))
vector = np.array([1, 2, 3])

# Think of four observations with three channels: add one offset per channel to every observation.
result = matrix + vector  # vector is broadcast across rows
print("Broadcasting 1D to 2D:")
print(result)

a = np.arange(3).reshape(3, 1)  # shape (3, 1)
b = np.arange(3)                # shape (3,)
print(f"\na shape: {a.shape}, b shape: {b.shape}")
print(f"a + b shape: {(a + b).shape}")
# The column a and row b produce a 3 by 3 table: entry [i, j] is a[i, 0] + b[j].
print(a + b)
```

### 2.5 How arrays are arranged in memory: strides


```python
# Strides give the spacing between stored entries in bytes along each dimension.
# Here the first stride moves to the next row and the second to the next column.
# Transposing changes these spacings without copying the numerical values.
arr = np.arange(12).reshape(3, 4)
print(f"Array:\n{arr}")
print(f"Shape: {arr.shape}")  # (rows, cols)

print(f"Strides: {arr.strides} bytes")
print(f"Item size: {arr.itemsize} bytes (each int64 is 8 bytes)")

# Swap the meanings of rows and columns; the underlying stored values stay in place.
arr_T = arr.T
print(f"\nTransposed strides: {arr_T.strides}")
print(f"Data is shared (no copy made): {np.shares_memory(arr, arr_T)}")
```

### 2.6 Einstein summation with einsum


```python
# einsum labels array indices: 'ij,jk->ik' gives matrix multiplication, summing over j.
# 'ii->' sums diagonal entries. The final example multiplies ten matrices by
# their corresponding vectors. Check the earlier results against @ and trace.
A = np.random.rand(3, 4)  # 3x4 matrix
B = np.random.rand(4, 5)  # 4x5 matrix

C1 = A @ B  # Result: 3x5 matrix

# For each output C[i, k], multiply A[i, j] by B[j, k] and sum over j. The letters before ->
# describe the inputs; those after it describe the output.
C2 = np.einsum('ij,jk->ik', A, B)

print(f"Results are equal: {np.allclose(C1, C2)}")

M = np.random.rand(5, 5)
trace1 = np.trace(M)
trace2 = np.einsum('ii->', M)  # Sum where both indices equal (diagonal)
print(f"Trace equal: {np.isclose(trace1, trace2)}")

batch = np.random.rand(10, 3, 4)  # 10 matrices of shape (3, 4)
vectors = np.random.rand(10, 4)    # 10 vectors of length 4

# Here i labels the observation, j the output component and k the input component. Sum over k,
# keeping a separate three-component result for each of the ten observations.
result = np.einsum('ijk,ik->ij', batch, vectors)
print(f"\nBatch multiply result shape: {result.shape}")  # (10, 3)
```

### 2.7 Apply a function to individual values or use an array expression


`np.vectorize` lets a function written for individual numbers accept arrays,
but it still runs that Python function once per pair of values. The second
expression uses NumPy's array operations directly.

`np.where(condition, first, second)` chooses between the two results at each
position. Both results are calculated before that choice, so it cannot protect
an invalid calculation such as division by zero. Use a row/value selection when
only some entries may safely be calculated.
[np.vectorize](https://numpy.org/doc/stable/reference/generated/numpy.vectorize)


```python
def python_function(x, y):
    return x - y if x > y else x + y

x_arr = np.array([1, 5, 3, 8])
y_arr = np.array([2, 3, 4, 1])
# Apply the scalar rule to corresponding pairs; otypes declares the result type. np.vectorize is a
# convenience wrapper, so it does not by itself make the Python function fast.
wrapped_result = np.vectorize(python_function, otypes=[np.int64])(x_arr, y_arr)
# Choose the difference where x > y and the sum elsewhere, using whole arrays.
native_result = np.where(x_arr > y_arr, x_arr - y_arr, x_arr + y_arr)
# Check that the two ways of expressing the rule give the same answers.
np.testing.assert_array_equal(wrapped_result, native_result)
print('Python function wrapped with np.vectorize:', wrapped_result)
print('Native array expression:', native_result)
```

## 3. Benchmarking loops and array operations


Here we calculate the same quantity using a Python loop and several pandas/NumPy
approaches, then compare their results and running times. Measuring this is called
**benchmarking**. The score is an example formula for this comparison, rather
than a physical model.

Calculating on whole arrays often saves the work of repeatedly entering a Python
loop. The actual time depends on the data size, number type, memory use and
computer. `np.vectorize` accepts arrays but still calls a Python function for each
value; its name does not imply a speed improvement.

These are single timing runs. For more reliable comparisons, use `%timeit` from
Section 0 and repeat the measurements. If a tool first compiles a function into
machine code, measure that preparation separately from later calls.


### 3.1 Create the table used for the timing comparison


Start with 100,000 rows so the comparison finishes quickly. Increasing `N`
lets you investigate how the time changes with data size. The inputs use 64-bit
integers (`int64`) so the products in this example fit within the stored number
range. Changing to a larger type after an overflow cannot recover the answer.


```python
N = 100_000
benchmark_rng = np.random.default_rng(0)
df_bench = pd.DataFrame({
    'a': benchmark_rng.integers(0, 1000, size=N, dtype=np.int64),
    'b': benchmark_rng.integers(1, 1000, size=N, dtype=np.int64),
})
```

```python
# Clean up any result columns from previous runs
# This ensures we get fresh timing measurements each time
result_cols = ["score_loop", "score_itertuples", "score_apply", "score_vec", "score_eval"]
for col in result_cols:
    if col in df_bench.columns:
        df_bench.drop(columns=[col], inplace=True)
```

```python
# Use the same integer formula for each timing method. % gives a remainder,
# so a % 3 == 0 tests divisibility by 3. This score is not a physical model.
def score_py(a: int, b: int) -> int:
    """An example integer formula for comparing calculation times."""
    return a * b + (a % 3 == 0) * b - (b % 5 == 0) * a
```

```python
RUN_SLOW_LOOP = False  # Enable to include individual DataFrame lookups.
timings = {}
```

### 3.2 Method 1: Explicit Python loop with DataFrame lookups


```python
# Loop over rows using .iat[row, column] to retrieve each value.
# This includes many separate table lookups. Enable RUN_SLOW_LOOP to time it.
if RUN_SLOW_LOOP:
    t0 = perf_counter()
    out_loop = np.empty(N, dtype=np.int64)  # reserve space for results

    for i in range(N):
        a = int(df_bench.iat[i, 0])
        b = int(df_bench.iat[i, 1])
        out_loop[i] = score_py(a, b)

    df_bench["score_loop"] = out_loop
    t1 = perf_counter()
    timings["Explicit for-loop"] = t1 - t0
```

### 3.3 Method 2: loop over rows with itertuples


```python
# itertuples(index=False, name=None) supplies each row as a tuple (a, b).
# Calculate the score for each pair and collect the answers into a column.
t2 = perf_counter()
out_tuples = [score_py(a, b) for a, b in df_bench[["a", "b"]].itertuples(index=False, name=None)]
df_bench["score_itertuples"] = np.array(out_tuples, dtype=np.int64)
t3 = perf_counter()
timings["itertuples + list comprehension"] = t3 - t2
```

### 3.4 Method 3: apply a function to each row


```python
# apply(..., axis=1) supplies one row at a time as a Series.
# The short lambda function extracts that row's a and b and calls score_py.
t4 = perf_counter()
df_bench["score_apply"] = df_bench.apply(lambda r: score_py(int(r.a), int(r.b)), axis=1)
t5 = perf_counter()
timings["DataFrame.apply(axis=1)"] = t5 - t4
```

### 3.5 Method 4: Native array operations


```python
# Calculate the score using whole columns. In this arithmetic,
# True contributes 1 and False contributes 0.
t6 = perf_counter()
df_bench["score_vec"] = (
    df_bench["a"] * df_bench["b"]                    # Base term: a*b
    + (df_bench["a"] % 3 == 0) * df_bench["b"]       # Bonus: +b if a divisible by 3
    - (df_bench["b"] % 5 == 0) * df_bench["a"]       # Penalty: -a if b divisible by 5
).astype(np.int64)
t7 = perf_counter()
timings["Vectorized ops"] = t7 - t6
```

### 3.6 Method 5: calculate a formula written as text with eval


```python
# eval reads a formula written as text, using column names as variables.
# inplace=True stores the resulting score column in the table.
t8 = perf_counter()
df_bench.eval("score_eval = a*b + (a % 3 == 0)*b - (b % 5 == 0)*a", inplace=True)
df_bench["score_eval"] = df_bench["score_eval"].astype(np.int64)  # Ensure int64 type
t9 = perf_counter()
timings["DataFrame.eval"] = t9 - t8
```

### 3.7 Verification and Results Comparison


```python
# Include every implementation that actually ran, including the optional slow loop.
checked_columns = []
for name in result_cols:
    if name in df_bench and name != 'score_vec':
        np.testing.assert_array_equal(df_bench['score_vec'], df_bench[name])
        checked_columns.append(name)
print('Verified against native array result:', ', '.join(checked_columns))
```

```python
timing_table = pd.DataFrame({'Elapsed time (s)': timings})
timing_table['Time / native array time'] = timing_table['Elapsed time (s)'] / timings['Vectorized ops']
display(timing_table.style.format({'Elapsed time (s)': '{:.4f}',
                                  'Time / native array time': '{:.2f}'}).set_caption(
    f'One run per method, N = {N:,}; ratios are measured on this machine'))
```

### 3.8 Interpreting the benchmark


First check that every method gives the same values, then compare times.
Use a whole-array expression when it expresses the calculation clearly.
`itertuples` supplies each row's values when a Python loop is needed.

`apply(axis=1)` still calls a Python function for each row. `eval` reads a formula
written as text, and its speed depends on how that formula is calculated on your
system. A command being part of pandas does not by itself make it faster. Repeat
the timing for the data size and calculation you actually need.

### 4.5 Pandas time parsing: dates, time zones, and durations


Before choosing a function, ask what the time value means. In an experiment,
**when an event occurred** and **how long after the trigger it occurred** are
different quantities. The same event can also have several written representations.

| Kind of value | Example | What it tells us | Suitable pandas representation |
|---|---|---|---|
| Text from a file | `'21/09/2026 09:00'` | Characters to interpret; the ordering and time zone must be known | Parse with `pd.to_datetime` |
| Local clock reading | `2026-09-21 10:00` in London | A date and wall-clock time; a zone is needed to locate an instant | Datetime followed by `tz_localize` |
| Instant on the civil-time timeline | `2026-09-21 09:00 UTC` | A particular event time with a reference time zone | A timezone-aware `Timestamp` |
| Elapsed duration | `1.5 s after the trigger` | A separation between times, with no calendar date of its own | `Timedelta` |

**UTC** means Coordinated Universal Time: a common reference for comparing clocks
in different places. Unlike local civil time in London, it does not switch between
summer and winter offsets. A time such as `09:00` by itself is incomplete: it does
not specify the date or time zone.

The familiar laboratory relationship is `event time = trigger time + elapsed duration`.
Subtracting two compatible timestamps gives a duration. Adding two timestamps is
not the operation we want. A column can hold many timestamps, while `pd.Timestamp`
represents one timestamp.


```python
# The trigger defines time zero for this experiment; UTC locates it on the calendar.
trigger_time = pd.Timestamp('2026-09-21 09:00:00', tz='UTC')
delay_after_trigger = pd.Timedelta(seconds=1.5)
event_time = trigger_time + delay_after_trigger
elapsed = event_time - trigger_time
display(pd.DataFrame({
    'Quantity': ['Trigger instant', 'Event instant', 'Elapsed duration'],
    'Value shown to the reader': [str(trigger_time), str(event_time), str(elapsed)],
    'Python/pandas type': [type(trigger_time).__name__, type(event_time).__name__, type(elapsed).__name__],
}))
print('Elapsed time for a numerical calculation:', elapsed.total_seconds(), 's')
assert elapsed.total_seconds() == 1.5
```

#### 4.5.1 Parsing text is interpreting a convention


A string is stored as text, not as a time coordinate. Parsing converts it into a
datetime value that supports time differences, sorting, and calendar operations.
`03/04/2026` means different dates under day/month/year and month/day/year conventions.

In `format='%d/%m/%Y %H:%M'`, `%d` is the day, `%m` the month, `%Y` the four-digit year,
`%H` the hour on a 24-hour clock, and `%M` the minute. **Case matters: `%m` and `%M`
mean different things.** Add `%S` for seconds and `%f` for fractional seconds.

`errors='coerce'` marks invalid inputs as **NaT (Not a Time)**, the datetime equivalent
of a missing measurement. Inspect these rows; a missing timestamp is not time zero.
The first example has no time zone yet, so it parses wall-clock readings without
claiming that they are UTC. Keep datetime values for calculations; format text for
display afterwards.
[Datetime parsing API](https://pandas.pydata.org/docs/reference/api/pandas.to_datetime.html)


```python
raw_dates = pd.Series(['03/04/2026 09:15', '21/09/2026 10:30', '31/02/2026 12:00', None])
parsed_dates = pd.to_datetime(raw_dates, format='%d/%m/%Y %H:%M', errors='coerce')
date_report = pd.DataFrame({'Input': raw_dates, 'Parsed (day/month/year)': parsed_dates,
                            'Invalid nonempty input': raw_dates.notna() & parsed_dates.isna()})
display(date_report)
print('March 4:', pd.to_datetime('03/04/2026', format='%m/%d/%Y'))
print('April 3:', pd.to_datetime('03/04/2026', format='%d/%m/%Y'))
assert parsed_dates.iloc[0] == pd.Timestamp('2026-04-03 09:15')
assert date_report['Invalid nonempty input'].sum() == 1
print('Before parsing:', raw_dates.dtype, '| after parsing:', parsed_dates.dtype)
# .dt exposes datetime operations for a Series. Formatting returns strings again.
display(pd.DataFrame({'Datetime for calculations': parsed_dates,
                       'Text for a report': parsed_dates.dt.strftime('%Y-%m-%d %H:%M')}))
```

#### 4.5.2 Explicit offsets and local clock times


In `2026-07-01T12:00:00+01:00`, `T` separates the date and time and `+01:00` means
that the displayed clock is one hour ahead of UTC. Subtract that offset to get
`11:00:00 UTC`. A final `Z` means UTC (an offset of zero). These ISO 8601 strings
describe instants without relying on an unstated local time zone.

- **Naive** means no time zone is attached. It does not mean UTC.
- **Aware** means a time zone or UTC offset is attached.
- `tz_localize` assigns a zone to an existing local clock reading: “this clock was in London”.
- `tz_convert` displays an already-known instant using a different zone: the event does not move.

Parse strings that already contain offsets with `utc=True` to compare their instants.
London uses different UTC offsets in summer and winter, so the zone name
`Europe/London` carries more information than a fixed `+01:00` offset. If the input
has no offset and was recorded in London, localize it there first, then convert to UTC.


```python
offset_strings = pd.Series(['2026-07-01T12:00:00+01:00', '2026-07-01T11:00:00Z'])
utc_instants = pd.to_datetime(offset_strings, format='ISO8601', utc=True)
display(pd.DataFrame({'Input': offset_strings, 'UTC': utc_instants,
                       'London': utc_instants.dt.tz_convert('Europe/London')}))
assert utc_instants.iloc[0] == utc_instants.iloc[1]

local_clocks = pd.to_datetime(pd.Series(['2026-07-01 12:00', '2026-12-01 12:00']),
                              format='%Y-%m-%d %H:%M')
london_instants = local_clocks.dt.tz_localize('Europe/London')
display(pd.DataFrame({'London': london_instants, 'UTC': london_instants.dt.tz_convert('UTC')}))
```

```python
# Daylight saving time (DST) changes the wall clock, not the sequence of instants.
# In spring, London's clock skips an hour: 01:30 on this date never occurs.
# In autumn, it repeats an hour: 01:30 on this date names two possible instants.
dst_clocks = pd.to_datetime(pd.Series(['2026-03-29 01:30', '2026-10-25 01:30']),
                           format='%Y-%m-%d %H:%M')
dst_checked = dst_clocks.dt.tz_localize('Europe/London', nonexistent='NaT', ambiguous='NaT')
display(pd.DataFrame({'Local clock': dst_clocks,
                       'Problem': ['Skipped clock reading', 'Repeated clock reading'],
                       'Unresolved instant': dst_checked}))
print('NaT means unresolved here; use an offset from the instrument or an explicit resolution policy.')
assert dst_checked.isna().all()
```

#### 4.5.3 Unix timestamps: a counter measured from an agreed starting instant


Think of a counter whose zero is **1970-01-01 00:00:00 UTC**. This agreed origin is
called the **Unix epoch**. A Unix timestamp conventionally counts seconds from that
origin. For example, `0` denotes the origin, `60` denotes one minute later, and `-1`
denotes one second before it. The choice of 1970 is a computing convention, not a
special date in the experiment. Logs on many operating systems use this convention;
it does not require using Unix yourself.

In the ordinary civil-time convention used here:

`represented time = reference origin + stored count × duration of one tick`

A bare number is insufficient: you need **both its unit and its origin**. An instrument
might count milliseconds since 1970, seconds since its last reboot, or sample numbers
since a trigger. Only the first is a millisecond encoding of Unix time. For sample
indices, first use the known sampling interval to obtain elapsed durations, then add
the trigger timestamp if it is known.

| `unit` argument | Duration of one tick | Ticks in one second |
|---|---|---:|
| `'s'` | second | 1 |
| `'ms'` | millisecond | 1,000 |
| `'us'` | microsecond (µs) | 1,000,000 |
| `'ns'` | nanosecond | 1,000,000,000 |

`pd.to_datetime(values, unit='s', origin='unix', utc=True)` tells pandas all three
things: the input counts are seconds, the origin is the Unix epoch, and the result
should be displayed in UTC. The input `unit` describes the numbers you were given;
changing it without rescaling those numbers changes the interpreted date.

Physics detail: Unix/POSIX time does not count leap seconds. It is not a substitute
for specifying a scientific time scale such as TAI when exact elapsed SI seconds
across leap seconds matter. [Python's definition](https://docs.python.org/3/library/time.html)


```python
# The same instants encoded using seconds and milliseconds since the Unix epoch.
epoch_seconds = pd.Series([-1, 0, 1, 60, 3600, 86400], dtype='int64')
epoch_milliseconds = epoch_seconds * 1000  # Convert the count, not the represented instant.
from_seconds = pd.to_datetime(epoch_seconds, unit='s', origin='unix', utc=True)
from_milliseconds = pd.to_datetime(epoch_milliseconds, unit='ms', origin='unix', utc=True)
display(pd.DataFrame({'Count since epoch (s)': epoch_seconds,
                       'Count since epoch (ms)': epoch_milliseconds,
                       'Readable date and time (UTC)': from_seconds}))
# Equal instants can use different storage resolutions. Compare at one resolution.
print('Seconds input produces:', from_seconds.dtype)
print('Milliseconds input produces:', from_milliseconds.dtype)
pd.testing.assert_series_equal(from_seconds.dt.as_unit('ms'), from_milliseconds)
```

#### 4.5.4 How pandas stores the time value


A normal pandas datetime column is not a column of printed date strings. Its values
are backed by **64-bit integer tick counts** together with a **resolution** such as
seconds or milliseconds. The dtype (data type) records how to interpret those counts.
Finer ticks use larger integers for the same date and reduce the date range that fits
in 64 bits. Read the dtype instead of assuming that every datetime column uses nanoseconds.

- `datetime64[ms]`: millisecond resolution, with no time zone. The values are naive
  clock readings; their numeric representation alone does not tell us the real UTC instant.
- `datetime64[ms, UTC]`: millisecond resolution with UTC attached.
- For a timezone-aware column, the underlying counts locate instants relative to UTC;
  the time zone is metadata used to produce the local calendar/clock representation.

A special missing-value encoding represents `NaT`. If missing values are present,
do not interpret every raw integer as a valid timestamp. The example below has no missing values.

**Storage resolution is not measurement accuracy.** Recording milliseconds as nanoseconds
adds no timing information. Clock calibration, synchronisation, and instrument uncertainty
still determine how accurately the event time is known. Converting to a coarser resolution
can lose fractional ticks; the example below only converts between exactly representable values.
[Pandas time representations](https://pandas.pydata.org/docs/user_guide/timeseries.html)


```python
storage_utc = pd.Series(pd.to_datetime([
    '2026-09-21T09:00:00.000Z', '2026-09-21T09:00:01.500Z',
], format='ISO8601', utc=True))
stored_ms = storage_utc.dt.as_unit('ms')
stored_ns = storage_utc.dt.as_unit('ns')
# astype('int64') exposes counts in the Series' CURRENT resolution.
ticks_ms = stored_ms.astype('int64')
ticks_ns = stored_ns.astype('int64')
display(pd.DataFrame({
    'UTC display': stored_ms,
    'London display (same instant)': stored_ms.dt.tz_convert('Europe/London'),
    'Stored count (ms since epoch)': ticks_ms,
    'Stored count (ns since epoch)': ticks_ns,
}))
print('Storage dtypes:', stored_ms.dtype, 'and', stored_ns.dtype)
restored_utc = pd.to_datetime(ticks_ms, unit='ms', origin='unix', utc=True)
pd.testing.assert_series_equal(restored_utc, stored_ms)
pd.testing.assert_series_equal(stored_ms.dt.as_unit('ns'), stored_ns)
pd.testing.assert_series_equal(stored_ms.dt.tz_convert('Europe/London').astype('int64'), ticks_ms)
print('Round trip restored the same instants; changing the display zone did not change the counts.')
```

#### 4.5.5 A unit mistake can produce a valid but completely wrong date


If a file supplies seconds but you tell pandas they are milliseconds, pandas has no
way to detect that mistake from the number alone. Check the file documentation and
whether the resulting dates make sense. Date magnitudes are clues, not a replacement
for knowing the instrument's format.


```python
# Obtain one example count in seconds from a known datetime.
example_unix_seconds = int(storage_utc.dt.as_unit('s').astype('int64').iloc[0])
display(pd.DataFrame({
    'Interpretation of the same integer': ['Correct: seconds', 'Incorrect: milliseconds'],
    'Input count': [example_unix_seconds, example_unix_seconds],
    'Resulting UTC timestamp': [
        pd.to_datetime(example_unix_seconds, unit='s', utc=True),
        pd.to_datetime(example_unix_seconds, unit='ms', utc=True),
    ],
}))
```

#### 4.5.6 Elapsed time: use the experiment's trigger as your origin


Subtracting timestamps gives a `Timedelta`, which stores a duration rather than a
calendar location. For fitting a decay or plotting a pulse, elapsed seconds from the
trigger are usually more useful than a large count since 1970. Subtract the trigger
while the data are still datetimes, then convert the resulting durations to seconds.

For a Series, `.dt.total_seconds()` returns the whole duration, including days and
fractional seconds. `.dt.seconds` returns only the seconds component within a day.
For one `Timedelta`, call `.total_seconds()` directly, without `.dt`.


```python
elapsed_from_trigger = storage_utc - storage_utc.iloc[0]
display(pd.DataFrame({'Event (UTC)': storage_utc, 'Time after first event (s)': elapsed_from_trigger.dt.total_seconds()}))
durations = pd.to_timedelta(pd.Series(['00:00:01.500', '1 days 00:00:02']))
display(pd.DataFrame({
    'Duration': durations,
    'Whole duration (s)': durations.dt.total_seconds(),
    'Seconds component only': durations.dt.seconds,
}))
assert elapsed_from_trigger.dt.total_seconds().iloc[1] == 1.5
assert durations.dt.total_seconds().iloc[1] == 86402
assert durations.dt.seconds.iloc[1] == 2
```

#### 4.5.7 Resampling and rolling measurements


`resample` groups observations into time bins; `rolling` calculates a moving window.
Keep missing readings visible and show the number of valid readings in each bin.
A trailing rolling window uses the past and current observation, while a centred
window would also use future measurements.


```python
time_rng = np.random.default_rng(42)
time_index = pd.date_range('2026-09-21 09:00', periods=25, freq='5min', tz='UTC')
time_measurements = pd.DataFrame({'temperature_C': 20 + np.linspace(0, 2, 25)
                                 + time_rng.normal(0, 0.15, 25)}, index=time_index)
time_measurements.index.name = 'time_utc'
time_measurements.iloc[5:7, 0] = np.nan
time_summary = time_measurements.resample('30min', closed='left', label='left').agg(
    mean_temperature_C=('temperature_C', 'mean'),
    valid_readings=('temperature_C', 'count'))
rolling_temperature = time_measurements['temperature_C'].rolling('20min', min_periods=2).mean()
display(time_summary.rename(columns={'mean_temperature_C': 'Mean temperature (°C)',
                                      'valid_readings': 'Valid readings (count)'}).style.format(
    {'Mean temperature (°C)': '{:.2f}'}, na_rep='—'))
fig, ax = plt.subplots(figsize=(9, 4), layout='constrained')
ax.plot(time_measurements.index, time_measurements['temperature_C'], 'o-', label='5 min readings')
ax.plot(rolling_temperature.index, rolling_temperature, label='Trailing 20 min mean')
ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M', tz=timezone.utc))
ax.set(xlabel='Time on 2026-09-21 (UTC)', ylabel='Temperature (°C)', title='Missing readings and a rolling mean')
ax.legend()
plt.show()
```

#### 4.5.8 Match each event to the latest available calibration


Match each event to a calibration recorded at or before it. Both tables
must first be sorted by the time column used for matching. With `merge_asof`,
`direction='backward'` selects earlier calibrations and `tolerance` sets the
maximum acceptable time difference. Keep the calibration time in the result
so you can check how old it was.

For several detectors, `by` also requires matching detector IDs. Use
`direction='nearest'` only if a calibration recorded after the event is acceptable.
[Matching by nearby times](https://pandas.pydata.org/docs/reference/api/pandas.merge_asof.html)


```python
event_times = pd.to_datetime(['2026-09-21 10:00:03', '2026-09-21 10:00:09',
                              '2026-09-21 10:00:30'], utc=True)
calibration_times = pd.to_datetime(['2026-09-21 10:00:00', '2026-09-21 10:00:08'], utc=True)
events = pd.DataFrame({'event_time': event_times, 'signal_V': [1.2, 1.4, 1.3]})
calibrations = pd.DataFrame({'calibration_time': calibration_times, 'temperature_C': [20.1, 20.3]})
aligned = pd.merge_asof(events.sort_values('event_time'), calibrations.sort_values('calibration_time'),
                        left_on='event_time', right_on='calibration_time',
                        direction='backward', tolerance=pd.Timedelta('5s'))
aligned['calibration_age_s'] = (aligned['event_time'] - aligned['calibration_time']).dt.total_seconds()
display(aligned.rename(columns={'signal_V': 'Signal (V)', 'temperature_C': 'Temperature (°C)',
                                 'calibration_age_s': 'Calibration age (s)'}))
assert aligned['temperature_C'].isna().sum() == 1
assert aligned['calibration_age_s'].dropna().between(0, 5).all()
```

## 5. pandas references and copies


Before modifying a table, check whether another variable refers to the same
table or to a separate result. These examples use **pandas 3** and its
**Copy-on-Write** behaviour: pandas can share stored data initially, but changing
a separate table or selected column does not change the original values.

- `alias = df` gives the same table another name. Editing `alias` also edits `df`.
- `s = df['B']` selects a separate Series. Editing `s` does not edit `df`.
- To change selected entries in `df`, use one `.loc[rows, columns] = value`
  operation. Two successive selections can instead change a temporary result.
- `.copy(deep=True)` copies the data immediately. `.copy(deep=False)` can wait
  until values are changed. If a cell contains a Python list or another mutable
  object, neither form automatically copies everything inside that object.

Run the examples in order and compare the original and modified values.
[pandas Copy-on-Write](https://pandas.pydata.org/docs/user_guide/copy_on_write.html)


### 5.1 Creating a Sample DataFrame


Create a small table whose values are easy to follow as we change them.


```python
df = pd.DataFrame({'A': [0, 1, 2, 3, 4], 'B': [10, 20, 30, 40, 50]})
display(df)
```

### 5.2 Two names for the same table


`df_alias = df` gives the same table a second name. Change a value through the new name and inspect `df`.

### 5.4 Why two successive selections do not update the table


This example deliberately uses two successive selections before assigning
a value. That does not update the original table in pandas 3. The warning is
captured and printed so you can see the problem; the following example uses `.loc`.


```python
# Demonstrate the mistake and capture its expected warning for explanation.
before_chained_assignment = df.copy()
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter('always', pd.errors.ChainedAssignmentError)
    df[df['A'] > 2]['B'] = -1
for warning in caught:
    print(f'{warning.category.__name__}: chained assignment cannot update df; use .loc.')
pd.testing.assert_frame_equal(df, before_chained_assignment)
display(df)
```

### 5.5 Change selected entries using loc


Select the rows and column together with `.loc`, then assign the new value to those entries.


```python
# Correct way: use .loc in one step
print("\nCorrect .loc assignment:")
df.loc[df["A"] > 2, "B"] = -1
print(df)
```

### 5.6 Make a separate copy of a table


Make a copy, edit it, and check the original. In pandas 3 both examples

### 6.1 Store missing values and inspect failed conversions


Some columns need to hold both numbers and missing entries. Pandas types
such as `Int64` and `Float64` allow this; the capital letters distinguish them
from some NumPy types. A **category** column stores repeated labels such as
detector names. When text cannot be converted to a number, keep track of that
failure instead of treating the missing reading as zero.


```python
import numpy as np
import pandas as pd
import io
from IPython.display import display

# Use a fixed seed so the artificial detector readings are repeatable.
pandas_rng = np.random.default_rng(210)
# Make one row per event; tile repeats the detector sequence A, B, C, D. Store signals as text
# first, as they might arrive from an instrument file.
raw_analysis_table = pd.DataFrame({
    'event_id': np.arange(24),
    'detector': np.tile(['A', 'B', 'C', 'D'], 6),
    'time': pd.date_range('2026-01-01', periods=24, freq='1min', tz='UTC'),
    'raw_signal': [f'{value:.3f}' for value in pandas_rng.normal(3, 0.8, 24)],
})
# Introduce two different problems: unreadable text and a measurement that is absent.
raw_analysis_table.loc[5, 'raw_signal'] = 'bad'
raw_analysis_table.loc[11, 'raw_signal'] = None
# assign adds or replaces columns and returns a new table. Each lambda is a short function whose
# frame argument is that table; Int64 and Float64 can store missing values.
analysis_table = raw_analysis_table.assign(
    event_id=lambda frame: frame['event_id'].astype('Int64'),
    detector=lambda frame: frame['detector'].astype('category'),
    signal_mV=lambda frame: pd.to_numeric(frame['raw_signal'], errors='coerce').astype('Float64'),
)
# Flag entries that contained text but became missing during conversion. An entry that was already
# absent is a different issue and is not selected here.
failed_numeric_conversion = analysis_table['raw_signal'].notna() & analysis_table['signal_mV'].isna()
# A missing signal gives an unknown comparison, rather than a definite False.
analysis_table['above_threshold'] = analysis_table['signal_mV'].gt(3).astype('boolean')
display(analysis_table.loc[failed_numeric_conversion, ['event_id', 'raw_signal']])
print(analysis_table.dtypes)
# For this selection only, treat an unknown comparison as not accepted; keep the original data.
print('Known positives:', analysis_table.loc[analysis_table['above_threshold'].fillna(False), 'event_id'].tolist())
```

### 6.2 Check matches when combining measurements and calibrations


A calibration table should contain one row for each detector ID. Several
measurements can then match that one calibration: this is a **many-to-one** match.
`validate='many_to_one'` checks for repeated calibration IDs, and `indicator=True`
shows which measurements found a match. Detector D deliberately has no calibration
here, so its calibrated measurement remains missing.


```python
detector_calibration = pd.DataFrame({'detector': ['A', 'B', 'C'],
                                     'gain': [1.0, 1.1, 0.9], 'offset_mV': [0.1, -0.2, 0.0]})
# Match calibration constants by detector name, keeping every measurement row. many_to_one allows
# repeated measurements but requires one calibration row per detector.
calibrated_table = analysis_table.merge(detector_calibration, on='detector', how='left',
                                        validate='many_to_one', indicator=True)
# Apply the gain and offset to each reading; missing calibration constants give a missing result.
calibrated_table['calibrated_mV'] = calibrated_table['signal_mV'] * calibrated_table['gain'] + calibrated_table['offset_mV']
# The added _merge column records whether each measurement found a calibration.
display(calibrated_table.groupby('_merge', observed=True).size())
print('Unmatched detectors:', calibrated_table.loc[calibrated_table['_merge'] == 'left_only', 'detector'].unique())
# Deliberately repeat detector A to show how an ambiguous calibration is detected.
duplicate_calibration = pd.concat([detector_calibration, detector_calibration.iloc[[0]]], ignore_index=True)
try:
    analysis_table.merge(duplicate_calibration, on='detector', validate='many_to_one')
# Catch this expected error so we can explain it and continue the example.
except pd.errors.MergeError as error:
    print('Caught invalid calibration keys:', error)
```

### 6.3 Calculate group summaries and compare individual readings


An **aggregation**, such as a grouped mean, produces one summary per group.
`transform` puts that group's summary beside each original row, making it easy
to subtract a detector's mean from its readings. `size` counts all rows; `count`
counts only the non-missing values. The ranking example also shows how equally
valued measurements are treated.


```python
# agg produces one row per detector. size counts all events; count excludes missing signals.
detector_summary = calibrated_table.groupby('detector', observed=True).agg(
    events=('event_id', 'size'), usable_signals=('calibrated_mV', 'count'),
    mean_mV=('calibrated_mV', 'mean'), spread_mV=('calibrated_mV', 'std'))
# transform repeats each detector mean beside its original readings, preserving row alignment.
group_means = calibrated_table.groupby('detector', observed=True)['calibrated_mV'].transform('mean')
calibrated_table['deviation_from_detector_mean_mV'] = calibrated_table['calibrated_mV'] - group_means
# Rank readings within their own detector; equal values share a rank, without gaps afterwards.
calibrated_table['rank_within_detector'] = calibrated_table.groupby('detector', observed=True)['calibrated_mV'].rank(
    method='dense', ascending=False)
display(detector_summary)
display(calibrated_table[['event_id', 'detector', 'calibrated_mV',
                          'deviation_from_detector_mean_mV', 'rank_within_detector']].head(8))
```

### 6.4 Reshape tables and check for repeated measurements


In a **wide** table, each channel has its own measurement column. In a
**long** table, a channel-name column identifies the measurement on each row.
The two channels here share the same unit.

`pivot` changes this layout and expects one reading for each row/column pair.
If a pair has several readings, decide how to summarise them with `pivot_table`,
or keep the repeats separate. Otherwise a layout change can hide repeated data.


```python
wide_measurements = pd.DataFrame({'sample_id': [101, 102, 103],
                                  'channel_A_mV': [1.2, 1.4, 1.3],
                                  'channel_B_mV': [2.1, 2.0, 2.2]})
# melt turns the two channel columns into rows, keeping sample_id with each reading.
long_measurements = wide_measurements.melt(id_vars='sample_id', var_name='channel', value_name='voltage_mV')
print('Duplicate sample/channel keys:', long_measurements.duplicated(['sample_id', 'channel']).any())
# pivot reverses this arrangement only when each sample/channel pair has one value.
restored_measurements = long_measurements.pivot(index='sample_id', columns='channel', values='voltage_mV')
display(long_measurements)
display(restored_measurements)

# Add a repeated measurement to demonstrate a case that needs a summary.
repeated_measurements = pd.concat([long_measurements, long_measurements.iloc[[0]]], ignore_index=True)
# pivot_table combines repeats using the chosen operations; count shows how many contributed.
repeat_summary = repeated_measurements.pivot_table(index='sample_id', columns='channel',
                                                   values='voltage_mV', aggfunc=['mean', 'count'])
display(repeat_summary)
```

### 6.5 Compute past-only rolling summaries within each detector

For each detector, compare a reading with the mean of the three previous
readings. Sort by detector and time first. `shift` moves the earlier values into
position before the rolling mean is calculated, excluding the current reading
from its own comparison. The window counts three observations, not three minutes.
Grouping ensures measurements from different detectors are not mixed.


```python
# Put each detector in time order before asking for its previous reading.
time_ordered_table = calibrated_table.sort_values(['detector', 'time']).copy()
# shift(1) moves readings down one place within each detector; its first row has no previous value.
time_ordered_table['previous_mV'] = time_ordered_table.groupby('detector', observed=True)['calibrated_mV'].shift(1)
# Shift before taking the rolling mean to exclude the current reading. The window uses up to three
# earlier rows from this detector; min_periods=1 allows one usable value.
time_ordered_table['past_three_mean_mV'] = time_ordered_table.groupby('detector', observed=True)['calibrated_mV'].transform(
    lambda series: series.shift(1).rolling(3, min_periods=1).mean())
time_ordered_table['change_since_previous_mV'] = time_ordered_table['calibrated_mV'] - time_ordered_table['previous_mV']
display(time_ordered_table[['event_id', 'detector', 'time', 'calibrated_mV',
                            'previous_mV', 'past_three_mean_mV']].head(10))
```

### 6.6 Flag unusual readings using the median and MAD


The **median absolute deviation (MAD)** is the median of the absolute
distances from a group's median. It gives a measure of spread that is less
affected by a few extreme readings than the standard deviation.

Use it here to flag readings for inspection. A zero MAD makes the division
undefined, so the score is left missing for that group. The scale factor used
below makes MAD comparable to a standard deviation for approximately normally
distributed data. A flag is a prompt to investigate, not a reason to delete a reading.


```python
# Give each reading its own detector median as a reference level.
grouped_medians = calibrated_table.groupby('detector', observed=True)['calibrated_mV'].transform('median')
absolute_deviation = (calibrated_table['calibrated_mV'] - grouped_medians).abs()
# MAD is the median absolute distance from that reference, calculated separately per detector.
grouped_mad = absolute_deviation.groupby(calibrated_table['detector'], observed=True).transform('median')
# For a Gaussian distribution, 1.4826 times MAD estimates the standard deviation. Replace a zero
# spread with a missing value so we do not divide by zero.
robust_scale_mV = (1.4826 * grouped_mad).replace(0, pd.NA)
calibrated_table['robust_deviation'] = (calibrated_table['calibrated_mV'] - grouped_medians) / robust_scale_mV
# Flag large deviations for inspection; the flag alone does not establish that a reading is wrong.
calibrated_table['needs_review'] = calibrated_table['robust_deviation'].abs().gt(3)
display(calibrated_table[['detector', 'calibrated_mV', 'robust_deviation', 'needs_review']].head(12))
```

### 6.7 Match the most recent calibration within a time tolerance

Attach the most recent earlier calibration for the same detector, using
`merge_asof`. Sort both tables by the time columns used for matching. `by` requires
the detector IDs to match too, and `tolerance` limits the calibration's age.
Keep the age and any missing matches visible so you can judge whether the
calibration is suitable.


```python
calibration_history = pd.DataFrame({
    'detector': ['A', 'B', 'C', 'A', 'B', 'C'],
    'calibration_time': pd.to_datetime(['2026-01-01 00:00Z'] * 3 + ['2026-01-01 00:12Z'] * 3),
    'historical_gain': [1.0, 1.1, 0.9, 1.02, 1.09, 0.92],
})
# Use the same text type for detector names in both tables before matching them.
timed_measurements = analysis_table.assign(detector=analysis_table['detector'].astype(str))
# Both time columns must be sorted. Match within a detector, using only earlier calibrations no more
# than five minutes old; unmatched measurements stay in the table.
matched_history = pd.merge_asof(timed_measurements.sort_values('time'),
                                calibration_history.sort_values('calibration_time'),
                                left_on='time', right_on='calibration_time', by='detector',
                                direction='backward', tolerance=pd.Timedelta('5min'))
# Subtract timestamps to get elapsed time, then express that duration in seconds.
matched_history['calibration_age_s'] = (matched_history['time'] - matched_history['calibration_time']).dt.total_seconds()
display(matched_history[['event_id', 'detector', 'time', 'historical_gain', 'calibration_age_s']].head(16))
print('Measurements without a recent matching calibration:', matched_history['historical_gain'].isna().sum())
```

### 6.8 Turn dictionaries and lists into rows while keeping event IDs


An event record may contain a dictionary of detector information and a
list of hits. `json_normalize` turns entries inside dictionaries into table
columns. `explode` makes one row per list item, keeping the event ID on each row.
An empty list produces a row with a missing item: keep it when counting events,
and explicitly exclude it when making a table of actual hits.


```python
nested_events = [
    {'event_id': 1, 'detector': {'name': 'A', 'gain': 1.1}, 'hit_energy_MeV': [0.4, 1.2]},
    {'event_id': 2, 'detector': {'name': 'B', 'gain': 0.9}, 'hit_energy_MeV': []},
    {'event_id': 3, 'detector': {'name': 'A', 'gain': 1.1}, 'hit_energy_MeV': [2.0]},
]
# Turn the nested detector information into columns such as detector_name.
flattened_events = pd.json_normalize(nested_events, sep='_')
# Make one row per hit while repeating its event information. An empty list leaves a missing hit.
exploded_events = flattened_events.explode('hit_energy_MeV', ignore_index=True)
# Keep only actual hits for calculations; retain flattened_events as the full event list.
hit_table = exploded_events.dropna(subset=['hit_energy_MeV']).copy()
hit_table['hit_energy_MeV'] = pd.to_numeric(hit_table['hit_energy_MeV'])
display(exploded_events)
# Count hits, then restore every original event ID so events with no hits receive a zero.
hit_counts = hit_table.groupby('event_id').size().reindex(flattened_events['event_id'], fill_value=0)
print('Hit counts, including events with no hits:')
display(hit_counts)
```

### 6.9 Choose bins by physical limits or numbers of observations


`cut` uses bin boundaries you specify, such as energy intervals in MeV.
`qcut` chooses boundaries to give groups with approximately equal numbers of
observations. These boundaries depend on the sample, so save them if they must
be reused. For model evaluation, choose any data-dependent boundaries from the
training data and then apply them unchanged to the test data.


```python
bin_values = analysis_table['signal_mV'].astype(float)
# cut uses limits chosen in measurement units; intervals include their
# right-hand boundary by default.
physical_signal_bins = pd.cut(bin_values, bins=[-np.inf, 0, 2, 4, np.inf],
                              labels=['Negative', 'Low', 'Nominal', 'High'])
# qcut instead chooses limits from these data to give roughly equal numbers of readings. If repeated
# values prevent distinct limits, duplicates="drop" allows fewer bins.
quantile_signal_bins, quantile_edges = pd.qcut(bin_values, q=4, retbins=True, duplicates='drop')
print('Physical bin counts:')
# Keep missing readings visible as a separate count when inspecting the bins.
display(physical_signal_bins.value_counts(sort=False, dropna=False))
print('Sample-derived quantile edges (mV):', quantile_edges)
display(quantile_signal_bins.value_counts(sort=False, dropna=False))
```

### 6.10 Read CSV data in chunks and calculate an overall mean


Read the file in **chunks**, each containing a limited number of rows.
To find an overall mean, add the sums and counts of valid measurements from all
chunks, then divide. Averaging the chunk means equally gives the wrong result
when the chunks contain different numbers of valid readings. `StringIO` holds
CSV text in memory for this example; a file path works with the same approach.


```python
csv_for_chunks = calibrated_table[['detector', 'calibrated_mV']].to_csv(index=False)
chunk_summaries = []
# StringIO lets this small text example act like a file; only five rows are read at a time.
with pd.read_csv(io.StringIO(csv_for_chunks), chunksize=5) as reader:
    for chunk in reader:
        # Keep the sum and the number of non-missing readings for each detector in this chunk.
        chunk_summaries.append(chunk.groupby('detector')['calibrated_mV'].agg(['sum', 'count']))
# The detector names are index labels: level=0 groups matching labels across the chunks.
combined_chunk_summary = pd.concat(chunk_summaries).groupby(level=0).sum()
# Overall mean = total signal / total count. Averaging chunk means would give small and large chunks
# equal weight, which can change the answer.
combined_chunk_summary['mean_mV'] = combined_chunk_summary['sum'] / combined_chunk_summary['count'].replace(0, np.nan)
full_table_means = calibrated_table.groupby('detector', observed=True)['calibrated_mV'].mean()
display(combined_chunk_summary)
# Compare with a direct calculation, first aligning detector order;
# matching missing values are allowed.
print('Agrees with full-table means:', np.allclose(
    combined_chunk_summary['mean_mV'].to_numpy(dtype=float),
    full_table_means.reindex(combined_chunk_summary.index).to_numpy(dtype=float, na_value=np.nan), equal_nan=True))
```

### 10.1 Keep analysis settings together in a dataclass


A **dataclass** groups related settings under one name, such as a detector
gain and a time offset. The type annotations show which kinds of values are
expected; they do not check those values automatically. `__post_init__` adds
the check used here. With `frozen=True`, settings cannot be reassigned; `replace`
creates a version with a changed setting while keeping the original. Units are
included in the names.


```python
from dataclasses import dataclass, asdict, replace
from collections import Counter, defaultdict
from itertools import batched, pairwise
from functools import lru_cache, partial
from concurrent.futures import ThreadPoolExecutor
from pathlib import Path
from tempfile import TemporaryDirectory
import csv
import io
import json
import math
import heapq
import re
import hashlib

# This line asks Python to build a settings container from the named fields below. frozen=True
# prevents reassignment; slots=True limits which attributes the container can have.
@dataclass(frozen=True, slots=True)
class AnalysisSettings:
    # The annotations describe the expected types; they do not check the values at runtime.
    threshold_mV: float = 2.0
    sample_period_s: float = 0.01
    detector_names: tuple[str, ...] = ('A', 'B')

    # Python calls this check after creating the settings: a sampling period must be positive.
    def __post_init__(self):
        if self.sample_period_s <= 0:
            raise ValueError('sample_period_s must be positive')

analysis_settings = AnalysisSettings()
# Make a new set of settings with one change, preserving the original experiment settings.
alternative_settings = replace(analysis_settings, threshold_mV=3.0)
# asdict converts the named settings to a dictionary that is easy to display or save.
print('Original:', asdict(analysis_settings))
print('Alternative:', asdict(alternative_settings))
```

### 10.2 Read text records and report conversion errors


Convert text to numbers when reading it. If a conversion fails, record
the line number and original text so you can inspect the input. `try` and
`except` handle the expected conversion error. A separate check rejects values
such as `nan`, which Python can read as floating-point values but which are not
usable measurements in this example.


```python
record_text = "detector,signal_mV\nA,1.2\nB,3.4\nA,bad\nB,nan\nA,2.5\n"
valid_records, rejected_records = [], []
# Read each row as a dictionary keyed by the CSV headings. Start at line 2 because line 1 contains
# the headings, not a measurement.
for line_number, row in enumerate(csv.DictReader(io.StringIO(record_text)), start=2):
    try:
        signal_mV = float(row['signal_mV'])
        # float accepts text such as "nan"; check separately for
        # missing or infinite numerical values.
        if not math.isfinite(signal_mV):
            raise ValueError('measurement is not finite')
    # Keep the line number and offending text so the original measurement can be investigated.
    except ValueError as error:
        rejected_records.append({'line': line_number, 'value': row['signal_mV'], 'reason': str(error)})
    # This else belongs to try: accept the record only if conversion and checks both succeeded.
    else:
        valid_records.append({'detector': row['detector'], 'signal_mV': signal_mV})
print('Accepted:', valid_records)
print('Rejected with context:', rejected_records)
```

### 10.3 Process measurements one at a time or in batches


A **generator** supplies one result at a time as the loop requests it.
It is used up by that loop; create a new generator to repeat the calculation.
`batched` groups items into manageable sets, with a possibly smaller final set.
The example keeps a running total rather than a list of every calculated value.


```python
# The * makes gain and offset_mV named arguments, which helps avoid swapping calibration constants.
def calibrated_signals(records, *, gain, offset_mV):
    for record in records:
        # yield supplies one calibrated reading, then pauses until the next value is requested.
        yield gain * record['signal_mV'] - offset_mV

signal_stream = calibrated_signals(valid_records, gain=1.1, offset_mV=0.2)
stream_total = 0.0
stream_count = 0
# Request two readings at a time; the last batch may contain only one.
for signal_batch in batched(signal_stream, 2):
    # Accumulate the sum and actual count, so a short final batch receives the right weight. fsum
    # reduces rounding error when adding floating-point values.
    stream_total += math.fsum(signal_batch)
    stream_count += len(signal_batch)
    print('Batch:', signal_batch)
print('Streaming mean (mV):', stream_total / stream_count)
# The stream has been used up; call calibrated_signals again to make a fresh one.
print('The consumed iterator now contains:', list(signal_stream))
```

### 10.4 Keep paired data aligned and compare neighbouring readings


`zip` pairs corresponding entries from several sequences. `strict=True`
raises an error if their lengths differ, catching a missing measurement or time.
`pairwise` supplies neighbouring entries, useful for time differences. Check
that the times are sorted before treating a difference as an elapsed interval.


```python
sample_times_s = [0.0, 0.1, 0.25, 0.4]
sample_values_mV = [1.0, 1.2, 1.5, 1.4]
# Keep each time paired with its measurement; strict=True rejects different input lengths.
aligned_samples = list(zip(sample_times_s, sample_values_mV, strict=True))
adjacent_slopes = []
# pairwise returns neighbouring pairs, and the nested brackets unpack their times and values.
for (t0, value0), (t1, value1) in pairwise(aligned_samples):
    if t1 <= t0:
        raise ValueError('Times must increase strictly')
    # This is a finite-difference estimate of dV/dt, using the actual time spacing for each pair.
    adjacent_slopes.append((value1 - value0) / (t1 - t0))
print('Local slopes (mV/s):', adjacent_slopes)
# Deliberately remove one value to demonstrate the alignment check.
try:
    list(zip(sample_times_s, sample_values_mV[:-1], strict=True))
except ValueError as error:
    print('Caught an intentional alignment error:', error)
```

### 10.5 Count categories and group records without a DataFrame

`Counter` counts repeated labels. `defaultdict(list)` creates a list for
each new label and collects the associated readings there. These are useful for
small collections of records; for numerical tables, compare them with pandas
grouping commands.


```python
detector_counts = Counter(record['detector'] for record in valid_records)
# Each new detector name automatically starts with an empty list of readings.
signals_by_detector = defaultdict(list)
for record in valid_records:
    signals_by_detector[record['detector']].append(record['signal_mV'])
# Build a dictionary containing one mean per detector, using only its own readings.
detector_means = {name: math.fsum(values) / len(values)
                  for name, values in signals_by_detector.items()}
print('Counts:', detector_counts)
print('Mean signals (mV):', detector_means)
```

### 10.6 Keep the largest events from a stream

### 10.7 Fix function arguments and reuse previous results


`partial` creates a version of a function with some arguments already
chosen, such as a fixed calibration gain. `lru_cache` remembers earlier answers
and reuses them when the inputs repeat; this is called **caching**.

Only cache a calculation whose result depends on its arguments and does not
change other data. Its arguments must work as dictionary keys, such as numbers
or strings; NumPy arrays generally do not. A limit on the cache size controls
how many earlier calls remain stored in memory.


```python
def convert_voltage(raw_counts, *, gain_mV_per_count, offset_mV):
    return raw_counts * gain_mV_per_count + offset_mV

# partial fixes detector A's calibration constants, leaving only raw_counts to supply later.
convert_detector_A = partial(convert_voltage, gain_mV_per_count=0.25, offset_mV=-0.5)
print('Calibrated voltage (mV):', convert_detector_A(20))

# Remember up to 32 previous input/result pairs. Repeated wavelengths can reuse their answer.
@lru_cache(maxsize=32)
def photon_energy_eV(wavelength_nm):
    if wavelength_nm <= 0:
        raise ValueError('Wavelength must be positive')
    # Use E = hc/lambda; hc is expressed in eV nm, matching the wavelength unit.
    return 1239.841984 / wavelength_nm

# The repeated 500 nm and 400 nm inputs demonstrate reuse of stored results.
photon_energies = [photon_energy_eV(value) for value in [400, 500, 600, 500, 400]]
print('Photon energies (eV):', photon_energies)
print('Cache statistics:', photon_energy_eV.cache_info())
```

### 10.9 Read several independent files with a thread pool


A **thread pool** lets several file-reading tasks be in progress together,
which can help while waiting for files. It does not guarantee faster numerical
loops. `executor.map` returns answers in the original file order, even when
files finish in another order. Each task returns its result to the main code.
These tiny files show the commands rather than measure a speed improvement.


```python
def file_mean(path):
    with path.open(encoding='utf-8') as source:
        # Ignore blank lines and convert the remaining text readings to numbers.
        values = [float(line) for line in source if line.strip()]
    # An empty file has no mean; stop with a message identifying the input.
    if not values:
        raise ValueError(f'No measurements in {path.name}')
    return {'file': path.name, 'mean': math.fsum(values) / len(values)}

# Make four small example files; real analyses could use a list of existing data paths.
with TemporaryDirectory(prefix='threaded-read-') as directory:
    input_paths = []
    for index in range(4):
        data_path = Path(directory) / f'part_{index}.txt'
        data_path.write_text(f'{index}\n{index + 1}\n{index + 2}\n')
        input_paths.append(data_path)
    # Allow two file reads to be in progress together while waiting for their contents.
    with ThreadPoolExecutor(max_workers=2) as executor:
        # map calls file_mean for each path and returns answers in input order; list collects them.
        concurrent_summaries = list(executor.map(file_mean, input_paths))
print(concurrent_summaries)
```

### 11.2 Add a progress bar to a loop and describe the work


Choose a description and unit that match the work, such as records or

```python
import numpy as np

progress_rng = np.random.default_rng(220)
progress_measurements = progress_rng.normal(3, 1, 1000)
accepted_measurement_count = 0
with tqdm(progress_measurements, desc='Check measurements', unit='record') as progress:
    for record_number, measurement in enumerate(progress, start=1):
        # Convert the threshold comparison to 1 for accepted or 0 for
        # rejected, then add it to the count.
        accepted_measurement_count += int(measurement >= 2.0)
        # The remainder is zero every 200 readings; update the status text at these intervals.
        if record_number % 200 == 0:
            progress.set_postfix(accepted=accepted_measurement_count, refresh=False)
print('Accepted records:', accepted_measurement_count)
```

### 11.5 Measure objects in an image stack with a progress indicator

Treat a stack of images as a sequence of frames. This loop selects bright
pixels and measures their total area in each frame, keeping the frame number
with each result. It is an image-processing loop, with no model training.

Revisit the example after Section 20 if image thresholds and masks are unfamiliar.


```python
import cv2
import numpy as np
import pandas as pd
from IPython.display import display

# Array axes are (frame, row, column); uint8 stores pixel intensities from 0 to 255.
progress_image_stack = np.zeros((20, 64, 64), dtype=np.uint8)
for frame_number, frame in enumerate(progress_image_stack):
    # Draw a filled disc into each frame, increasing its radius by one pixel per frame.
    cv2.circle(frame, (32, 32), 5 + frame_number, 180, thickness=-1)
image_stack_results = []
for frame_number, frame in enumerate(tqdm(progress_image_stack, desc='Measure image stack', unit='frame')):
    # Keep pixels brighter than 100. The mask contains 255 for selected pixels and 0 elsewhere.
    _, foreground = cv2.threshold(frame, 100, 255, cv2.THRESH_BINARY)
    # Count selected pixels for the area, then average the original intensity only in that region.
    # cv2.mean returns channel means as a tuple; [0] takes the one grayscale channel.
    image_stack_results.append({'frame': frame_number, 'area_pixels': cv2.countNonZero(foreground),
                                'mean_foreground_intensity': cv2.mean(frame, mask=foreground)[0]})
image_stack_table = pd.DataFrame(image_stack_results)
display(image_stack_table.head())
print('Frames measured:', len(image_stack_table))
```

### 11.6 Report progress when the total is not known in advance

Some sources supply records without announcing the final number. Leave
`total` unset: tqdm then reports the count, elapsed time and processing rate,
but cannot calculate a percentage or remaining time. There is no need to read
the whole source into a list just to count it first.


```python
def measurement_records():
    for record_number in range(73):
        # Supply one record at a time, as an instrument stream might do.
        yield {'index': record_number, 'signal': record_number / 10}

streamed_record_count = 0
streamed_signal_total = 0.0
# No total is supplied, so the display reports a count and rate rather than a percentage.
with tqdm(desc='Summarise stream', unit='record') as unknown_total_progress:
    for record in measurement_records():
        streamed_record_count += 1
        streamed_signal_total += record['signal']
        # Advance only after this record has contributed to the running totals.
        unknown_total_progress.update(1)
print('Final stream count:', streamed_record_count)
print('Signal total:', streamed_signal_total)
```

### 11.7 Update progress by bytes while processing a stream

For data read in blocks, count the number of **bytes** processed: the last
block may be shorter than the others. `update(len(block))` adds the amount just
read. `unit_scale=True` displays readable prefixes. The checksum example compares
a digest of the blocks with one calculated from the complete input.


```python
import hashlib
import io

binary_payload = bytes(range(256)) * 257
streamed_digest = hashlib.sha256()
bytes_processed = 0
# Use the byte sequence as a readable stream, without creating a file on disk.
with io.BytesIO(binary_payload) as byte_stream:
    with tqdm(total=len(binary_payload), desc='Checksum input', unit='B',
               unit_scale=True, unit_divisor=1024) as byte_progress:
        # := stores the next block and tests whether it is non-empty. An empty block ends the loop.
        while block := byte_stream.read(4096):
            # Feed each block into the same running checksum, in its original order.
            streamed_digest.update(block)
            bytes_processed += len(block)
            # Advance by bytes actually read; the final block may contain fewer than 4096 bytes.
            byte_progress.update(len(block))
print('Bytes processed:', bytes_processed)
print('Digest:', streamed_digest.hexdigest())
# Verify that processing blocks gives the same digest as processing the whole input at once.
print('Matches direct checksum:', streamed_digest.digest() == hashlib.sha256(binary_payload).digest())
```

### 11.8 Use nested bars for files and chunks

The outer bar counts files; the inner bar counts chunks in the current file.
`leave=False` removes completed inner bars to avoid a long trail of output.
`tqdm.write` prints a message without disrupting text progress bars. Keep each
total in the same units as the work counted by its loop.


```python
import numpy as np
import pandas as pd
from IPython.display import display

nested_chunk_summaries = []
for file_number in tqdm(range(3), desc='Files', unit='file', position=0):
    # Split 103 readings into five chunks; their lengths differ
    # because 103 is not divisible by five.
    file_chunks = np.array_split(np.arange(103) + file_number, 5)
    # Use a second display line for the inner bar and remove it when this file is finished.
    for chunk in tqdm(file_chunks, desc=f'File {file_number}: chunks', unit='chunk',
                       position=1, leave=False):
        # Retain the file number so each chunk contributes to the correct file summary.
        nested_chunk_summaries.append({'file': file_number, 'count': len(chunk), 'sum': chunk.sum()})
tqdm.write('Finished all files and chunks.')
# Add chunk counts and sums within each file to recover the totals for that file.
nested_file_summary = pd.DataFrame(nested_chunk_summaries).groupby('file').agg(
    rows=('count', 'sum'), signal_total=('sum', 'sum'))
display(nested_file_summary)
```

### 11.9 Track completed file reads from a thread pool

Each submitted file-reading task has a **future**, which provides its result
when ready. `as_completed` supplies these as they finish, so the bar tracks
completed reads. `future.result()` returns the answer or raises the error from
that task. Keep file names and sort the results afterwards if input order matters.
The example demonstrates overlapping file reads, rather than a speed benchmark.


```python
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
from tempfile import TemporaryDirectory
import json
import pandas as pd
from IPython.display import display

# Each task reads one file and returns its name, number of readings and total signal.
def summarise_json_file(path):
    values = json.loads(path.read_text())
    return {'file': path.name, 'count': len(values), 'sum': sum(values)}

completed_file_summaries = []
# Make eight practice files and remove them automatically after all reads have finished.
with TemporaryDirectory(prefix='tqdm-files-') as directory:
    progress_paths = []
    for file_number in range(8):
        path = Path(directory) / f'run_{file_number:02d}.json'
        path.write_text(json.dumps(list(range(file_number, file_number + 20))))
        progress_paths.append(path)
    # Allow up to three file-reading tasks to be in progress together.
    with ThreadPoolExecutor(max_workers=3) as executor:
        # submit starts a task and returns a future: an object from
        # which we can retrieve its answer.
        pending_reads = [executor.submit(summarise_json_file, path) for path in progress_paths]
        # as_completed supplies tasks in finishing order, so the bar tracks completed reads.
        for future in tqdm(as_completed(pending_reads), total=len(pending_reads),
                            desc='Read data files', unit='file'):
            # result retrieves the returned dictionary, or reports an error
            # raised by that file-reading task.
            completed_file_summaries.append(future.result())
# Restore filename order because tasks may have finished in a different order.
completed_file_summaries.sort(key=lambda row: row['file'])
display(pd.DataFrame(completed_file_summaries))
```

### 12.5 Displaying units on tables and plots


Keep numeric values numeric and document their units. Convert the values before
changing a unit label: `current_A * 1000` produces milliamperes. Formatting a value
with an `mA` suffix alone does not convert it. pandas column names, `attrs`, and
Matplotlib labels are descriptive metadata; they do not check dimensional consistency.

The examples use standard Matplotlib text/MathText and pandas styling, with no

### 14.2 Pie and donut charts with counts and percentages

### 14.3 Measurement uncertainties and a residual panel

Compare generated voltage measurements with a supplied exponential curve.
The curve's parameters are fixed. Error bars show the known measurement standard
deviation; the shaded band shows that same noise scale around the curve, rather
than uncertainty in fitted parameters.

A **residual** is the measurement minus the predicted value. Plotting residuals
separately helps reveal differences hidden by the overall trend. Dividing by the
measurement uncertainty gives a dimensionless residual.


```python
voltage_rng = np.random.default_rng(21)
voltage_time_s = np.linspace(0, 5, 35)
# Use a known exponential decay as the reference signal, with a time constant of 1.4 s.
expected_voltage_V = 3.0 * np.exp(-voltage_time_s / 1.4)
voltage_sigma_V = np.full_like(voltage_time_s, 0.12)
# Add independent Gaussian noise with standard deviation sigma to simulate measurements.
observed_voltage_V = expected_voltage_V + voltage_rng.normal(0, voltage_sigma_V)
# Use a taller signal panel and a shorter residual panel, sharing the same time axis.
fig, axes = plt.subplots(2, 1, figsize=(7, 6), sharex=True,
                         gridspec_kw={'height_ratios': [3, 1]}, layout='constrained')
axes[0].errorbar(voltage_time_s, observed_voltage_V, yerr=voltage_sigma_V,
                 fmt='o', ms=4, capsize=2, label='Measurements')
axes[0].plot(voltage_time_s, expected_voltage_V, label='Supplied curve')
# Shade the supplied noise scale around the curve; this is not a fitted confidence interval.
axes[0].fill_between(voltage_time_s, expected_voltage_V - voltage_sigma_V,
                     expected_voltage_V + voltage_sigma_V, alpha=0.2,
                     label='Known noise scale: ±1 standard deviation')
axes[0].set(ylabel='Voltage (V)', title='Signal and residuals')
axes[0].legend(fontsize=8)
# Divide each observed-minus-expected difference by its uncertainty. The result is dimensionless: a
# value of 2 means two noise standard deviations above the curve.
normalised_residuals = (observed_voltage_V - expected_voltage_V) / voltage_sigma_V
# Each uncertainty is now one in these normalised units.
axes[1].errorbar(voltage_time_s, normalised_residuals, yerr=np.ones(35), fmt='.', capsize=2)
axes[1].axhline(0, color='black', lw=1)
axes[1].set(xlabel='Time (s)', ylabel='Residual / σ')
plt.show()
```

### 14.4 Logarithmic axes and ranges spanning orders of magnitude

Logarithmic axes require positive values. Use `semilogy` for an exponential

### 14.5 Two-dimensional histograms and hexagonal bins

Thousands of overlapping points can hide the density of observations. Binning
makes local counts visible. Here colour represents **events per bin**, not a
probability density. A logarithmic colour scale reveals both sparse tails and
the dense centre; empty bins are omitted from that logarithmic display.


```python
density_rng = np.random.default_rng(22)
# Generate 6000 (x, y) observations; the off-diagonal covariance creates a tilted cloud.
density_points_mm = density_rng.multivariate_normal([0, 0], [[4, 1.5], [1.5, 2]], 6000)
fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout='constrained')
# .T turns the coordinate columns into rows, and * passes those rows as separate x and y inputs.
# LogNorm gives colour a logarithmic count scale; cmin=1 hides empty bins.
_, _, _, hist_image = axes[0].hist2d(*density_points_mm.T, bins=40,
                                   norm=LogNorm(), cmin=1, cmap='viridis')
# Count the same events in hexagonal bins; mincnt=1 again excludes empty bins.
hex_image = axes[1].hexbin(*density_points_mm.T, gridsize=35, mincnt=1,
                          norm=LogNorm(), cmap='viridis')
# Match each panel to its own coloured plot object so its colour bar describes the correct counts.
for ax, artist, title in zip(axes, [hist_image, hex_image], ['Rectangular bins', 'Hexagonal bins']):
    ax.set(xlabel='x (mm)', ylabel='y (mm)', title=title, aspect='equal')
    fig.colorbar(artist, ax=ax, label='Events per bin')
plt.show()
```

### 14.6 Contours and heatmaps of a scalar field

`meshgrid` constructs coordinate arrays; `potential_V[y_index, x_index]` stores
the field value. Contours join equal values. `pcolormesh` displays the field
across the plane and also supports nonuniform grids. Contour labels and the colour bar use the same voltage units.


```python
field_x_m = np.linspace(-2, 2, 65)
field_y_m = np.linspace(-1.5, 1.5, 49)
# Build coordinates at every grid point. With indexing="xy", rows follow y and columns follow x.
field_X, field_Y = np.meshgrid(field_x_m, field_y_m, indexing='xy')
potential_V = 5.0 * np.exp(-(field_X**2 + 2 * field_Y**2))
fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout='constrained')
# Draw lines at the listed potential values; each number is in volts.
contours = axes[0].contour(field_X, field_Y, potential_V, levels=[0.5, 1, 2, 3, 4], colors='black')
axes[0].clabel(contours, fmt='%g V', fontsize=8)
# Colour each part of the same field, with a fixed 0 to 5 V scale.
field_image = axes[1].pcolormesh(field_X, field_Y, potential_V,
                                shading='auto', cmap='viridis', vmin=0, vmax=5)
for ax, title in zip(axes, ['Equipotential contours', 'Potential field']):
    ax.set(xlabel='x (m)', ylabel='y (m)', title=title, aspect='equal')
fig.colorbar(field_image, ax=axes[1], label='Potential (V)')
plt.show()
```

### 14.7 Vector fields with arrows and streamlines

For rigid rotation at 1 rad/s, the velocity is `v_x = -y` and `v_y = x` in
metres per second. Arrows show local directions and magnitudes; streamlines
show curves tangent to the field. Sample fewer arrows to prevent overlap.


```python
flow_axis_m = np.linspace(-2, 2, 25)
flow_X, flow_Y = np.meshgrid(flow_axis_m, flow_axis_m)
# Specify a circulating flow: v_x = -omega*y and v_y = omega*x, with omega = 1 s^-1.
velocity_x = -flow_Y
velocity_y = flow_X
# hypot calculates sqrt(v_x**2 + v_y**2), the speed used to colour the plot.
speed_m_s = np.hypot(velocity_x, velocity_y)
fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout='constrained')
# [::2, ::2] keeps every second grid point in both directions so the arrows remain readable.
arrows = axes[0].quiver(flow_X[::2, ::2], flow_Y[::2, ::2],
                        velocity_x[::2, ::2], velocity_y[::2, ::2], speed_m_s[::2, ::2],
                        cmap='viridis', angles='xy', scale_units='xy', scale=6)
axes[0].quiverkey(arrows, 0.85, 1.05, 1, '1 m/s', labelpos='E')
# Streamlines follow the local velocity direction; their colour shows speed, not elapsed time.
streams = axes[1].streamplot(flow_axis_m, flow_axis_m, velocity_x, velocity_y,
                             color=speed_m_s, cmap='viridis', density=1.1)
for ax, title in zip(axes, ['Velocity vectors', 'Streamlines']):
    ax.set(xlabel='x (m)', ylabel='y (m)', title=title, aspect='equal')
fig.colorbar(streams.lines, ax=axes[1], label='Speed (m/s)')
plt.show()
```

### 14.8 Three-dimensional trajectories and point clouds

A line joins positions in trajectory order; a scatter plot shows separate
points. Colour can show another quantity such as time. Label all three spatial
axes. Points can hide behind one another in a 3D view, so inspect several angles
or compare 2D projections. Section 15 includes controls for rotating the view.


```python
trajectory_time_s = np.linspace(0, 4, 150)
track_x_cm = np.cos(2 * np.pi * trajectory_time_s)
track_y_cm = np.sin(2 * np.pi * trajectory_time_s)
track_z_cm = 0.5 * trajectory_time_s
track_rng = np.random.default_rng(23)
# Put x, y and z in columns, keeping each row as one position at one time.
track_points = np.column_stack([track_x_cm, track_y_cm, track_z_cm])
# Add independent position noise with standard deviation 0.04 cm to each coordinate.
measured_track = track_points + track_rng.normal(0, 0.04, track_points.shape)
fig = plt.figure(figsize=(10, 4), layout='constrained')
# Create two plotting areas with a third spatial axis; i + 1 selects panel 1 or 2.
track_axes = [fig.add_subplot(1, 2, i + 1, projection='3d') for i in range(2)]
track_axes[0].plot(track_x_cm, track_y_cm, track_z_cm)
# Unpack the three coordinate columns; use the matching time values to colour the points.
track_scatter = track_axes[1].scatter(*measured_track.T, c=trajectory_time_s, s=8, cmap='viridis')
for ax, title in zip(track_axes, ['Ordered trajectory', 'Measured positions coloured by time']):
    ax.set(xlabel='x (cm)', ylabel='y (cm)', zlabel='z (cm)', title=title)
    ax.set_box_aspect((2, 2, 2))
fig.colorbar(track_scatter, ax=track_axes[1], label='Time (s)', shrink=0.65)
plt.show()
```

### 14.11 Box plots and violin plots with individual measurements

A box plot shows the median and middle half of the values. Its whiskers
and separately marked points help show the remaining spread. A violin plot
uses width to show a smoothed estimate of the distribution. Adding individual
points makes the sample size and original readings visible. The violin width
describes the spread of measurements, not the uncertainty in their mean.


```python
groups_rng = np.random.default_rng(25)
# Generate 35 readings for each detector, allowing different mean signals and spreads.
detector_signals_mV = [groups_rng.normal(mean, width, 35)
                       for mean, width in [(10, 1), (12, 1.8), (11, 0.6)]]
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True, layout='constrained')
# The box shows the middle half of the readings; the line inside marks the median.
axes[0].boxplot(detector_signals_mV, tick_labels=['A', 'B', 'C'])
axes[1].violinplot(detector_signals_mV, showmedians=True)
# Place individual observations over the violin at detector positions 1, 2 and 3.
for index, samples in enumerate(detector_signals_mV, start=1):
    # Add a small random horizontal offset to separate overlapping
    # points; signal values stay unchanged.
    axes[1].scatter(index + groups_rng.uniform(-0.06, 0.06, len(samples)), samples,
                     s=9, alpha=0.6, color='black')
axes[1].set_xticks([1, 2, 3], ['A', 'B', 'C'])
for ax, title in zip(axes, ['Quartile summary', 'Distribution and observations']):
    ax.set(xlabel='Detector', ylabel='Signal (mV)', title=title)
plt.show()
```

### 14.12 Contours from irregularly spaced measurements

`tricontourf` joins irregularly spaced measurement positions into triangles

it cannot reveal unmeasured fine detail. Show the measurement positions too,
especially where they are far apart.


```python
irregular_rng = np.random.default_rng(26)
probe_x_m = irregular_rng.uniform(-2, 2, 150)
probe_y_m = irregular_rng.uniform(-2, 2, 150)
probe_voltage_V = np.exp(-(probe_x_m**2 + probe_y_m**2))
fig, ax = plt.subplots(figsize=(6, 5), layout='constrained')
# Join the irregular probe positions into triangles to draw contours between measurements. The
# smooth-looking coloured regions do not add new measured information.
irregular_field = ax.tricontourf(probe_x_m, probe_y_m, probe_voltage_V,
                                 levels=12, cmap='viridis')
# Overlay the actual sample positions so readers can see where observations support the map.
ax.scatter(probe_x_m, probe_y_m, s=5, color='black', label='Probe positions')
ax.set(xlabel='x (m)', ylabel='y (m)', title='Irregularly sampled field', aspect='equal')
ax.legend()
fig.colorbar(irregular_field, ax=ax, label='Voltage (V)')
plt.show()
```

### 14.13 Take slices in three directions through an array


The array is stored as `volume[z, y, x]`. Choosing a single index along one
dimension gives a 2D slice through it. Label the two remaining directions in
the same order as the displayed array. Use one shared colour scale so that a
given colour represents the same value in every panel.


```python
volume_axis_mm = np.linspace(-3, 3, 31)
# Choose array axis order (z, y, x); indexing="ij" preserves that order.
volume_Z, volume_Y, volume_X = np.meshgrid(volume_axis_mm, volume_axis_mm, volume_axis_mm, indexing='ij')
density_volume = np.exp(-(volume_X**2 + 2 * volume_Y**2 + 0.5 * volume_Z**2))
# Integer division finds the central sample; on this grid it is coordinate 0 mm.
middle_index = len(volume_axis_mm) // 2
# Fix one index and keep all values along the other axes. These slices have row/column orders (y,
# x), (z, x) and (z, y), respectively.
volume_slices = [density_volume[middle_index, :, :],
                 density_volume[:, middle_index, :],
                 density_volume[:, :, middle_index]]
volume_labels = [('x (mm)', 'y (mm)', 'z = 0 mm'),
                  ('x (mm)', 'z (mm)', 'y = 0 mm'),
                  ('y (mm)', 'z (mm)', 'x = 0 mm')]
fig, axes = plt.subplots(1, 3, figsize=(11, 3.5), layout='constrained')
# Share one colour scale so the same colour means the same density in all three panels.
volume_norm = Normalize(vmin=0, vmax=1)
# Unpack a matching plot, slice and set of labels together on each pass through the loop.
for ax, values, (xlabel, ylabel, title) in zip(axes, volume_slices, volume_labels):
    slice_artist = ax.pcolormesh(volume_axis_mm, volume_axis_mm, values,
                                  shading='auto', cmap='magma', norm=volume_norm)
    ax.set(xlabel=xlabel, ylabel=ylabel, title=title, aspect='equal')
fig.colorbar(slice_artist, ax=axes, label='Relative density (dimensionless)', shrink=0.8)
plt.show()
```

### 14.14 Add histograms beside a scatter plot


`subplot_mosaic` lets us give names to panels in a figure. The main panel
shows pairs of measurements; the side histograms show the separate distributions
of x and y. Sharing the relevant axes keeps these views aligned when limits change.


```python
marginal_rng = np.random.default_rng(27)
marginal_points = marginal_rng.multivariate_normal([0, 0], [[1, 0.7], [0.7, 1]], 800)
# Name the plotting areas with a small layout map; the dot leaves the upper-right space empty.
fig, panels = plt.subplot_mosaic([['top', '.'], ['main', 'right']],
                                  width_ratios=[4, 1], height_ratios=[1, 4],
                                  figsize=(6, 6), layout='constrained')
# Link each histogram coordinate to the corresponding scatter axis for direct comparison.
panels['top'].sharex(panels['main'])
panels['right'].sharey(panels['main'])
panels['main'].scatter(*marginal_points.T, s=7, alpha=0.25)
panels['top'].hist(marginal_points[:, 0], bins=25)
# Put y along the vertical axis and counts along the horizontal axis in the right-hand histogram.
panels['right'].hist(marginal_points[:, 1], bins=25, orientation='horizontal')
panels['main'].set(xlabel='x (mm)', ylabel='y (mm)')
panels['top'].set_ylabel('Events')
panels['right'].set_xlabel('Events')
# Hide repeated tick labels while keeping the shared scales.
panels['top'].tick_params(labelbottom=False)
panels['right'].tick_params(labelleft=False)
fig.suptitle('A joint distribution and its projections')
plt.show()
```

### 15.2 Explore amplitude, frequency and damping with sliders

The sliders change amplitude, frequency and damping in a supplied formula.
The curve is recalculated when a setting changes; no parameters are fitted to
measurements. The axis limits stay fixed to make amplitude changes easy to see.
`interactive_output` connects each control to the matching function argument.


```python
# Keep all drawing in a function so the same calculation works with sliders or a direct call.
def draw_oscillation(amplitude, frequency, damping):
    time_s = np.linspace(0, 5, 500)
    # Compute A exp(-gamma*t) cos(2*pi*f*t): frequency is in Hz and damping is in s^-1.
    displacement_cm = amplitude * np.exp(-damping * time_s) * np.cos(2 * np.pi * frequency * time_s)
    fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
    ax.plot(time_s, displacement_cm)
    ax.set(xlabel='Time (s)', ylabel='Displacement (cm)', ylim=(-3.2, 3.2),
           title='A supplied damped-oscillation model')
    ax.grid(alpha=0.3)
    plt.show()
    # Release the figure after displaying it so repeated slider changes
    # do not accumulate open figures.
    plt.close(fig)

if ENABLE_WIDGETS:
    # Each slider supplies one function argument.
    # continuous_update=False waits until you release it.
    amplitude_control = widgets.FloatSlider(value=1, min=0, max=3, step=0.1,
                                             description='A (cm)', continuous_update=False)
    frequency_control = widgets.FloatSlider(value=1, min=0.2, max=3, step=0.1,
                                             description='f (Hz)', continuous_update=False)
    damping_control = widgets.FloatSlider(value=0.3, min=0, max=1, step=0.05,
                                           description='γ (s⁻¹)', continuous_update=False)
    # Map argument names to controls; changing a control calls draw_oscillation with the new values.
    wave_output = widgets.interactive_output(draw_oscillation, {
        'amplitude': amplitude_control, 'frequency': frequency_control, 'damping': damping_control,
    })
    # VBox arranges controls vertically; wave_output holds the plot produced by the function.
    display(widgets.VBox([amplitude_control, frequency_control, damping_control]), wave_output)
else:
    # Produce a normal static plot when widget controls are disabled.
    draw_oscillation(amplitude=1, frequency=1, damping=0.3)
```

### 15.3 Change histogram bins, selection range and normalisation

The generated measurements stay the same while you change the controls.
Changing the bins changes the display; choosing a smaller range also excludes
observations. If density is selected, the histogram of the remaining observations
has area one and units inverse to those on the horizontal axis.


```python
histogram_sample_V = np.random.default_rng(31).normal(0, 1, 1200)

def draw_histogram(bins, interval, density):
    # Unpack the selected voltage interval into its two endpoints.
    low, high = interval
    # Use both comparisons together to retain readings within that interval.
    selected = histogram_sample_V[(histogram_sample_V >= low) & (histogram_sample_V <= high)]
    fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
    # A histogram needs some selected readings and an interval of nonzero width.
    if len(selected) and high > low:
        # density=True scales the histogram so its area is one; heights then have units of V^-1.
        ax.hist(selected, bins=bins, range=(low, high), density=density,
                edgecolor='white', color='tab:blue')
    else:
        # transAxes uses coordinates from 0 to 1 across the plot,
        # placing this message in its centre.
        ax.text(0.5, 0.5, 'No finite-width selection to plot', transform=ax.transAxes, ha='center')
    ax.set(xlabel='Voltage (V)', ylabel='Density (V⁻¹)' if density else 'Events per bin',
           xlim=(-4, 4), title=f'{len(selected)} of {len(histogram_sample_V)} observations selected')
    plt.show()
    plt.close(fig)

if ENABLE_WIDGETS:
    bins_control = widgets.IntSlider(value=25, min=5, max=80, description='Bins', continuous_update=False)
    interval_control = widgets.FloatRangeSlider(value=(-3, 3), min=-4, max=4, step=0.1,
                                                 description='Range (V)', continuous_update=False)
    density_control = widgets.Checkbox(value=False, description='Probability density')
    # Connect bins, interval and density to the arguments of the drawing function.
    histogram_output = widgets.interactive_output(draw_histogram, {
        'bins': bins_control, 'interval': interval_control, 'density': density_control,
    })
    display(widgets.VBox([bins_control, interval_control, density_control]), histogram_output)
else:
    draw_histogram(bins=25, interval=(-3, 3), density=False)
```

### 15.4 Choose scatter axes and apply an event threshold

Use the menus to choose the horizontal and vertical quantities. An energy
threshold selects events; the same selection is applied to every coordinate so
each point still represents one event. The colour scale continues to show energy
over the same fixed range.


```python
explorer_rng = np.random.default_rng(32)
# Each array contains 400 entries in the same event order, so index i
# refers to the same event throughout.
explorer_data = {
    'x (mm)': explorer_rng.normal(0, 2, 400),
    'y (mm)': explorer_rng.normal(0, 1, 400),
    'z (mm)': explorer_rng.normal(0, 3, 400),
    'Energy (MeV)': explorer_rng.uniform(0, 5, 400),
}

def draw_event_scatter(horizontal, vertical, minimum_energy):
    # Build one True/False selection from the energy cut and reuse it for every plotted quantity.
    accepted = explorer_data['Energy (MeV)'] >= minimum_energy
    fig, ax = plt.subplots(figsize=(6, 4), layout='constrained')
    # horizontal and vertical are dictionary keys chosen by the controls. Keep the colour scale at 0
    # to 5 MeV so colours retain their meaning as the cut changes.
    points = ax.scatter(explorer_data[horizontal][accepted], explorer_data[vertical][accepted],
                        c=explorer_data['Energy (MeV)'][accepted], vmin=0, vmax=5,
                        s=16, alpha=0.6, cmap='viridis')
    ax.set(xlabel=horizontal, ylabel=vertical, title=f'{accepted.sum()} accepted events')
    fig.colorbar(points, ax=ax, label='Energy (MeV)')
    plt.show()
    plt.close(fig)

if ENABLE_WIDGETS:
    # Offer the dictionary keys as axis choices, including the units in their displayed names.
    horizontal_control = widgets.Dropdown(options=list(explorer_data), value='x (mm)', description='Horizontal')
    vertical_control = widgets.Dropdown(options=list(explorer_data), value='y (mm)', description='Vertical')
    threshold_control = widgets.FloatSlider(value=0, min=0, max=5, step=0.1,
                                             description='Min MeV', continuous_update=False)
    # Use the control values as named arguments whenever the plotting function is called.
    scatter_output = widgets.interactive_output(draw_event_scatter, {
        'horizontal': horizontal_control, 'vertical': vertical_control, 'minimum_energy': threshold_control,
    })
    display(widgets.VBox([horizontal_control, vertical_control, threshold_control]), scatter_output)
else:
    draw_event_scatter(horizontal='x (mm)', vertical='y (mm)', minimum_energy=0)
```

### 15.5 Rotate a three-dimensional view using camera controls

Elevation is the viewing angle above or below the horizontal plane; azimuth
turns the view around the vertical axis. Changing these angles redraws the figure
without changing the trajectory's coordinates. Inspect several views to check
whether an apparent crossing or bend is a viewing effect.


```python
def draw_camera(elevation, azimuth):
    phase_rad = np.linspace(0, 6 * np.pi, 160)
    fig = plt.figure(figsize=(6, 4), layout='constrained')
    ax = fig.add_subplot(projection='3d')
    ax.plot(np.cos(phase_rad), np.sin(phase_rad), phase_rad / (3 * np.pi))
    # Change the viewing direction in degrees; the helix coordinates themselves stay the same.
    ax.view_init(elev=elevation, azim=azimuth)
    ax.set(xlabel='x (cm)', ylabel='y (cm)', zlabel='z (cm)',
           title=f'Elevation {elevation}°, azimuth {azimuth}°')
    ax.set_box_aspect((2, 2, 2))
    plt.show()
    plt.close(fig)

if ENABLE_WIDGETS:
    elevation_control = widgets.IntSlider(value=25, min=-90, max=90, description='Elevation', continuous_update=False)
    azimuth_control = widgets.IntSlider(value=35, min=-180, max=180, description='Azimuth', continuous_update=False)
    # Call the drawing function with the elevation and azimuth selected by the two controls.
    camera_output = widgets.interactive_output(draw_camera, {'elevation': elevation_control, 'azimuth': azimuth_control})
    display(widgets.HBox([elevation_control, azimuth_control]), camera_output)
else:
    draw_camera(elevation=25, azimuth=35)
```

### 15.6 Select an axis and move through a three-dimensional volume

The array order is `(z, y, x)`. The dropdown chooses which dimension is held
fixed; the slider chooses an index on that dimension. `np.take` expresses this
selection directly. Keeping the colour range fixed prevents each slice from
appearing equally bright when its actual maximum changes.


```python
slice_coordinates_mm = np.linspace(-3, 3, 31)
# Store the volume in (z, y, x) axis order, matching the array selections below.
slice_Z, slice_Y, slice_X = np.meshgrid(slice_coordinates_mm, slice_coordinates_mm,
                                      slice_coordinates_mm, indexing='ij')
interactive_volume = np.exp(-((slice_X - 0.7)**2 + 2 * slice_Y**2 + 0.5 * slice_Z**2))

def draw_volume_slice(direction, index):
    # Translate the chosen physical direction into its array axis number.
    axis_number = {'z': 0, 'y': 1, 'x': 2}[direction]
    # Fix one index along that axis; np.take returns the remaining two-dimensional plane.
    values = np.take(interactive_volume, index, axis=axis_number)
    # Label the surviving axes in plotting order: horizontal column, then vertical row.
    xlabel, ylabel = {'z': ('x', 'y'), 'y': ('x', 'z'), 'x': ('y', 'z')}[direction]
    fig, ax = plt.subplots(figsize=(6, 4), layout='constrained')
    # Use the same 0 to 1 colour scale for every slice to compare densities fairly.
    image = ax.pcolormesh(slice_coordinates_mm, slice_coordinates_mm, values,
                          shading='auto', vmin=0, vmax=1, cmap='magma')
    ax.set(xlabel=f'{xlabel} (mm)', ylabel=f'{ylabel} (mm)', aspect='equal',
           title=f'{direction} = {slice_coordinates_mm[index]:.1f} mm; array shape {values.shape}')
    fig.colorbar(image, ax=ax, label='Relative density (dimensionless)')
    plt.show()
    plt.close(fig)

if ENABLE_WIDGETS:
    direction_control = widgets.Dropdown(options=['z', 'y', 'x'], description='Fixed axis')
    # The slider chooses an array index from 0 to 30; the plot title
    # converts it to a position in mm.
    slice_control = widgets.IntSlider(value=15, min=0, max=30, description='Slice index', continuous_update=False)
    slice_output = widgets.interactive_output(draw_volume_slice, {'direction': direction_control, 'index': slice_control})
    display(widgets.HBox([direction_control, slice_control]), slice_output)
else:
    draw_volume_slice(direction='z', index=15)
```

### 15.7 Connect an event cut to a pie or donut chart

Move the energy threshold to see how many events pass a selection. Counts
always refer to the same 500 events. The switch changes only the chart style;
it does not change the selected data.


```python
acceptance_energy_MeV = np.random.default_rng(33).uniform(0, 5, 500)

def draw_acceptance(threshold, donut):
    # Count events passing the energy cut; every other event belongs to the rejected group.
    accepted_count = np.count_nonzero(acceptance_energy_MeV >= threshold)
    counts = [accepted_count, len(acceptance_energy_MeV) - accepted_count]
    fig, ax = plt.subplots(figsize=(5, 4), layout='constrained')
    # Convert these two counts into fractions of the original 500 events.
    ax.pie(counts, labels=['Accepted', 'Rejected'], autopct='%1.1f%%', startangle=90,
            colors=['#0072B2', '#E69F00'],
            # Choose a narrow ring for a donut or a full-radius wedge for an ordinary pie chart.
            wedgeprops={'width': 0.4 if donut else 1.0, 'edgecolor': 'white'})
    ax.set(aspect='equal', title=f'{counts[0]} accepted, {counts[1]} rejected; cut ≥ {threshold:.1f} MeV')
    plt.show()
    plt.close(fig)

if ENABLE_WIDGETS:
    acceptance_control = widgets.FloatSlider(value=2, min=0, max=5, step=0.1,
                                              description='Cut (MeV)', continuous_update=False)
    donut_control = widgets.Checkbox(value=True, description='Donut style')
    # Recalculate both counts and redraw whenever the threshold or chart-style control changes.
    acceptance_output = widgets.interactive_output(draw_acceptance, {'threshold': acceptance_control, 'donut': donut_control})
    display(widgets.HBox([acceptance_control, donut_control]), acceptance_output)
else:
    draw_acceptance(threshold=2, donut=True)
```

### 15.8 Embed an animation with playback and frame controls

`FuncAnimation` updates an existing line, avoiding a new figure for every frame.
`to_jshtml` embeds the frames and browser playback controls, so this example needs
neither a video encoder nor a live kernel after rendering. More frames or larger
figures increase the saved notebook size. Here 36 small frames show a travelling
wave. The horizontal coordinate is position; the changing coordinate is time.
[Matplotlib animations](https://matplotlib.org/stable/gallery/animation/index.html)


```python
animation_x_m = np.linspace(0, 2, 150)
# Choose 36 physical times for the wave; omit the endpoint to avoid repeating the first phase.
animation_times_s = np.linspace(0, 1, 36, endpoint=False)
animation_fig, animation_ax = plt.subplots(figsize=(5, 3), dpi=80, layout='constrained')
# plot returns a list of lines; the trailing comma takes its single line object.
animation_line, = animation_ax.plot(animation_x_m, np.sin(2 * np.pi * animation_x_m))
animation_ax.set(xlabel='Position (m)', ylabel='Displacement (cm)', ylim=(-1.2, 1.2))
animation_title = animation_ax.set_title('Time = 0.00 s')

def update_wave_frame(frame_index):
    time_s = animation_times_s[frame_index]
    # Update the existing line with sin(2*pi*(x - v*t)), using wavelength 1 m and speed 1 m/s.
    animation_line.set_ydata(np.sin(2 * np.pi * (animation_x_m - time_s)))
    animation_title.set_text(f'Time = {time_s:.2f} s')
    return animation_line, animation_title

# Call update_wave_frame for each frame index. interval is playback time in milliseconds, which need
# not equal the physical time between the chosen samples.
wave_animation = FuncAnimation(animation_fig, update_wave_frame,
                                frames=len(animation_times_s), interval=60, blit=False)
# Embed the frames and playback controls as HTML, without requiring live widget controls.
wave_animation_html = wave_animation.to_jshtml(default_mode='loop')
plt.close(animation_fig)
display(HTML(wave_animation_html))
```

### 18.2 Find the minimum of a function


Numerical **optimisation** searches for values that minimise or maximise
a function. Here `minimize` starts from a supplied guess and seeks the minimum
of a two-variable test function. Inspect the returned success flag and function
value. In a physical calculation you would also check whether the result lies
in the allowed range of your parameters.


```python
# Optimization - minimize a function
def rosenbrock(x):
    """Rosenbrock function - has minimum at (1, 1)"""
    return (1 - x[0])**2 + 100*(x[1] - x[0]**2)**2

# Initial guess
x0 = [0, 0]

# Minimize
result = minimize(rosenbrock, x0, method='BFGS')

print(f"Optimization result:")
print(f"Optimal point: {result.x}")
print(f"Function value: {result.fun:.6f}")
print(f"Success: {result.success}")
```

### 19.2 Integrate a formula and unevenly sampled measurements


For a decaying current $I(t)=I_0e^{-t/\tau}$, integrating over time gives charge.
`quad` takes a Python function for the current and chooses evaluation points
to estimate the integral. It also returns an estimate of numerical error.
`simpson` instead takes measurements already stored in an array; supply their
actual time coordinates, particularly when the sampling is uneven.

Neither numerical method calculates measurement uncertainty. Compare with the
analytic charge over $0\leq t\leq T$, which is $I_0\tau(1-e^{-T/\tau})$.
[Integration commands](https://docs.scipy.org/doc/scipy/reference/integrate.html)


```python
charge_I0_A, charge_tau_s, charge_stop_s = 2.0, 0.7, 4.0

def decay_current_A(time_s):
    return charge_I0_A * np.exp(-time_s / charge_tau_s)

# Charge is the area under I(t): integrating amperes over seconds gives coulombs. quad chooses its
# own evaluation times to meet the requested numerical tolerances.
charge_quad_C, charge_error_C = integrate.quad(decay_current_A, 0, charge_stop_s,
                                               epsabs=1e-10, epsrel=1e-10)
# Use deliberately uneven sample times to mimic measurements with irregular spacing.
charge_times_s = charge_stop_s * np.linspace(0, 1, 101)**1.4
# Pass the actual time coordinates to Simpson integration, rather than assuming equal spacing.
charge_sampled_C = integrate.simpson(decay_current_A(charge_times_s), x=charge_times_s)
# Use the analytic integral of this exponential as a check on both numerical methods.
charge_exact_C = charge_I0_A * charge_tau_s * (1 - np.exp(-charge_stop_s / charge_tau_s))
display(pd.DataFrame({'method': ['quad', 'sampled Simpson', 'analytic'],
                      'charge_C': [charge_quad_C, charge_sampled_C, charge_exact_C]}))
# This estimates integration error for the supplied function, not uncertainty in measured currents.
print(f'quad estimated numerical error: {charge_error_C:.2e} C')
print(f'Sampled integration discrepancy: {abs(charge_sampled_C - charge_exact_C):.2e} C')
np.testing.assert_allclose(charge_quad_C, charge_exact_C, rtol=1e-10)
```

### 19.3 Solve a damped oscillator and locate zero crossings


Write the oscillator equation using position and velocity together:
$y=[x,v]$, $dx/dt=v$ and $dv/dt=-\omega_0^2x-2\gamma v$.
The function supplied to `solve_ivp` returns these two rates of change at the
current time and state. The solution has shape **(state components, time samples)**:
its first row contains positions and its second contains velocities.

An **event function** identifies times when a chosen quantity is zero. Here it
finds downward crossings of $x=0$. Limiting the step size helps avoid missing
crossings. `t_eval` gives the times to store, while the solver chooses its own
intermediate steps. `dense_output=True` also allows values to be calculated
between the stored times.

The tolerances set targets for the numerical error: absolute tolerances use the
units of each state component; relative tolerance is dimensionless. Check the
returned `success` flag and compare the solution with the known formula.
[solve_ivp](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html)


```python
osc_omega0_rad_s, osc_gamma_per_s = 2 * np.pi, 0.35

# Rewrite x'' + 2*gamma*x' + omega0^2*x = 0 as two first-order equations. The solver stores state =
# [position, velocity] and needs [velocity, acceleration] in return.
def oscillator_rhs(time_s, state):
    position_m, velocity_m_s = state
    return [velocity_m_s, -osc_omega0_rad_s**2 * position_m - 2 * osc_gamma_per_s * velocity_m_s]

# An event function reports zero when the event occurs; here it returns the position.
def downward_zero(time_s, state):
    return state[0]

# Record only crossings from positive to negative displacement.
downward_zero.direction = -1
osc_times_s = np.linspace(0, 5, 501)
# Start at x = 0.1 m and v = 0 m/s, and request output at osc_times_s. The solver chooses internal
# steps; max_step limits their size to help resolve crossings. dense_output provides interpolation
# between outputs; atol lists position and velocity tolerances.
osc_solution = integrate.solve_ivp(oscillator_rhs, (0, 5), [0.1, 0.0],
    t_eval=osc_times_s, events=downward_zero, dense_output=True, max_step=0.05,
    rtol=1e-8, atol=[1e-10, 1e-9])
# Check that the numerical solver finished successfully before using its answer.
if not osc_solution.success:
    raise RuntimeError(osc_solution.message)
# The result has two rows (position and velocity) and one column per requested time.
osc_x_m, osc_v_m_s = osc_solution.y
# Calculate the damped frequency for this underdamped case, then compare with its analytic solution.
osc_wd_rad_s = np.sqrt(osc_omega0_rad_s**2 - osc_gamma_per_s**2)
osc_exact_m = 0.1 * np.exp(-osc_gamma_per_s * osc_times_s) * (
    np.cos(osc_wd_rad_s * osc_times_s) + osc_gamma_per_s / osc_wd_rad_s * np.sin(osc_wd_rad_s * osc_times_s))
np.testing.assert_allclose(osc_x_m, osc_exact_m, atol=2e-9)
print('State array shape:', osc_solution.y.shape)
# t_events contains one array per event function, so [0] selects our downward crossings.
print('Downward crossings (s):', np.round(osc_solution.t_events[0], 3))
print('Interpolated [x (m), v (m/s)] at 1.234 s:', osc_solution.sol(1.234))
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5), layout='constrained')
axes[0].plot(osc_times_s, osc_x_m)
axes[0].scatter(osc_solution.t_events[0], np.zeros(len(osc_solution.t_events[0])), color='tab:red', label='Events')
axes[0].set(xlabel='Time (s)', ylabel='Displacement (m)', title='Damped oscillator')
axes[0].legend()
# A phase-space plot pairs position and velocity at the same time, showing the decaying motion.
axes[1].plot(osc_x_m, osc_v_m_s)
axes[1].set(xlabel='Displacement (m)', ylabel='Velocity (m/s)', title='Phase-space trajectory')
axes[1].locator_params(axis='x', nbins=5)
plt.show()
plt.close(fig)
```

### 19.4 Find when an equation reaches a chosen value


Find the time when an object, cooling towards room temperature with a known
time constant, reaches 40 °C. Subtract the target temperature from the formula
so that the question becomes finding a **root**, where $f(t)=0$.

`brentq` needs two times enclosing the root, called a **bracket**. For a continuous
function, opposite signs at these times show that a root lies between them.
Choose a physically meaningful interval and check `converged` in the result.
The equation and its parameters are supplied; no parameters are fitted to data.
[root_scalar](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.root_scalar.html)


```python
cool_room_C, cool_initial_C, cool_target_C, cool_tau_s = 20.0, 90.0, 40.0, 30.0

def cooling_residual_C(time_s):
    # Subtract the target temperature so reaching it becomes the equation residual = 0.
    return cool_room_C + (cool_initial_C - cool_room_C) * np.exp(-time_s / cool_tau_s) - cool_target_C

# The residual changes sign between 0 and 200 s; brentq searches within that bracket.
cool_result = optimize.root_scalar(cooling_residual_C, bracket=(0, 200), method='brentq', xtol=1e-10)
if not cool_result.converged:
    raise RuntimeError(cool_result.flag)
# Rearrange the exponential cooling law analytically to check the numerical crossing time.
cool_exact_s = cool_tau_s * np.log((cool_initial_C - cool_room_C) / (cool_target_C - cool_room_C))
np.testing.assert_allclose(cool_result.root, cool_exact_s, atol=1e-9)
print(f'Threshold reached after {cool_result.root:.3f} s ({cool_result.iterations} iterations)')
# Evaluate the original equation at the returned time; the difference should be close to zero.
print(f'Temperature residual: {cooling_residual_C(cool_result.root):.2e} °C')
```

### 19.5 Solve a heat-flow equation with a sparse matrix


A uniformly heated rod with both ends fixed at 20 °C satisfies
$-d^2T/dx^2=q/k$. Approximate the second derivative using neighbouring grid
points. Most matrix entries are zero: this is a **sparse** matrix, with only
three nonzero diagonals. `diags_array` stores those diagonals,
and `spsolve` solves $AT=b$ without forming a dense inverse. Boundary values
contribute to the first and last entries of the right-hand side.

The analytic solution is $T(x)=20+(q/k)x(L-x)/2$. The residual checks the
linear solve; comparison with a known solution also checks the discretisation.
For this quadratic solution the finite-difference formula is exact apart from
rounding. For a general solution, repeat with more closely spaced grid points
to check how much the answer changes.
[Sparse arrays](https://docs.scipy.org/doc/scipy/reference/sparse.html)


```python
rod_length_m, rod_source_K_m2, rod_boundary_C = 1.0, 80.0, 20.0
rod_n = 99  # Interior points; endpoints are supplied boundary conditions.
rod_dx_m = rod_length_m / (rod_n + 1)
rod_x_m = np.linspace(0, rod_length_m, rod_n + 2)
# For -d²T/dx² = source, the finite-difference row is (-T_left + 2*T_here - T_right)/dx². Only three
# diagonals are needed; the sparse representation avoids storing the other zeros.
rod_operator = sparse.diags_array([-np.ones(rod_n - 1), 2 * np.ones(rod_n), -np.ones(rod_n - 1)],
                                  offsets=[-1, 0, 1], format='csc') / rod_dx_m**2
rod_rhs = np.full(rod_n, rod_source_K_m2)
# Move the known end temperatures to the right-hand side of the first and last interior equations.
rod_rhs[[0, -1]] += rod_boundary_C / rod_dx_m**2
# Solve the simultaneous equations A @ T = b for the unknown interior temperatures.
rod_interior_C = sparse_linalg.spsolve(rod_operator, rod_rhs)
# np.r_ joins the left boundary, interior solution and right boundary into the complete profile.
rod_temperature_C = np.r_[rod_boundary_C, rod_interior_C, rod_boundary_C]
# For this constant source and equal end temperatures, the analytic profile is a parabola.
rod_exact_C = rod_boundary_C + rod_source_K_m2 * rod_x_m * (rod_length_m - rod_x_m) / 2
np.testing.assert_allclose(rod_temperature_C, rod_exact_C, atol=1e-10)
print('Matrix shape:', rod_operator.shape, '; stored entries:', rod_operator.nnz)
# Substitute the solution back into the equations; a small residual checks the numerical solve.
print('Relative linear residual:', np.linalg.norm(rod_operator @ rod_interior_C - rod_rhs) / np.linalg.norm(rod_rhs))
fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
ax.plot(rod_x_m, rod_exact_C, label='Analytic')
ax.plot(rod_x_m[::5], rod_temperature_C[::5], 'o', label='Sparse solve')
ax.set(xlabel='Position (m)', ylabel='Temperature (°C)', title='Steady temperature along a rod')
ax.legend()
plt.show()
plt.close(fig)
```

### 19.6 Calculate normal modes using a generalised eigenproblem


For coupled masses, $K u=\omega^2 M u$, where $K$ is the symmetric stiffness
matrix and $M$ is a positive-definite mass matrix. `eigh(K, M)` uses that
structure directly. Eigenvectors are columns; their overall sign is arbitrary.
The returned modes are orthonormal in the mass-weighted inner product,
$U^T M U=I$. Dividing angular frequency by $2\pi$ gives frequency in hertz.
[eigh](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.eigh.html)


```python
# Describe three masses coupled by springs between fixed walls; M contains masses and K stiffnesses.
mode_mass_kg = np.diag([1.0, 1.5, 1.0])
mode_stiffness_N_m = 20.0 * np.array([[2, -1, 0], [-1, 2, -1], [0, -1, 2]], dtype=float)
# Solve K u = omega² M u. Each returned column is one mode shape; each eigenvalue is omega².
mode_omega_squared, mode_vectors = linalg.eigh(mode_stiffness_N_m, mode_mass_kg)
# Take the square root to obtain angular frequency, then divide by 2*pi to get cycles per second.
mode_frequency_Hz = np.sqrt(mode_omega_squared) / (2 * np.pi)
# Check the solver's mass-weighted normalisation, then check the original mode equation.
np.testing.assert_allclose(mode_vectors.T @ mode_mass_kg @ mode_vectors, np.eye(3), atol=1e-12)
np.testing.assert_allclose(mode_stiffness_N_m @ mode_vectors,
                           (mode_mass_kg @ mode_vectors) * mode_omega_squared, atol=1e-12)
display(pd.DataFrame({'mode': [1, 2, 3], 'frequency_Hz': mode_frequency_Hz}))
fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
for mode_index in range(3):
    # Rescale each mode for display so its largest displacement has magnitude one.
    relative_shape = mode_vectors[:, mode_index] / np.max(np.abs(mode_vectors[:, mode_index]))
    # Add zero displacement at the two fixed walls, either side of the three masses.
    ax.plot(range(5), np.r_[0, relative_shape, 0], 'o-', label=f'Mode {mode_index + 1}')
ax.set(xlabel='Position index (0 and 4 are fixed walls)', ylabel='Relative displacement', title='Normal-mode shapes')
ax.set_xticks(range(5))
ax.legend()
plt.show()
plt.close(fig)
```

### 19.7 Measure frequencies with a correctly scaled real FFT


`rfft` keeps the non-negative frequencies of a real signal. `rfftfreq` needs
the sampling interval in seconds to return hertz. For this **even-length**
record, a one-sided amplitude spectrum doubles interior bins but leaves the DC
and Nyquist bins unchanged. For an odd-length record, the last bin must also be
doubled because there is no Nyquist bin.

The sinusoids here complete an integer number of cycles, so their amplitudes
fall on FFT bins. General signals have spectral leakage; windowing changes
amplitude scaling. Uniform sampling and a suitable sampling rate are essential.
[FFT tutorial](https://docs.scipy.org/doc/scipy/tutorial/fft.html)


```python
fft_sample_rate_Hz, fft_count = 200.0, 400
fft_time_s = np.arange(fft_count) / fft_sample_rate_Hz
# Combine a 0.3 V constant offset with known 20 Hz and 55 Hz oscillations.
fft_voltage_V = 0.3 + 1.2 * np.sin(2 * np.pi * 20 * fft_time_s) + 0.4 * np.cos(2 * np.pi * 55 * fft_time_s)
# rfftfreq supplies the frequencies for real-valued data; d is the sampling interval in seconds.
fft_frequency_Hz = fft.rfftfreq(fft_count, d=1 / fft_sample_rate_Hz)
# Divide Fourier magnitudes by the sample count to recover amplitudes in volts.
fft_amplitude_V = np.abs(fft.rfft(fft_voltage_V)) / fft_count
# For this even sample count, double the positive-frequency amplitudes except DC and Nyquist. Those
# two endpoints do not have separate negative-frequency partners.
fft_amplitude_V[1:-1] *= 2
# Both tones fall exactly on frequency bins here, so compare directly with the supplied amplitudes.
np.testing.assert_allclose(fft_amplitude_V[[0, 40, 110]], [0.3, 1.2, 0.4], atol=1e-12)
print(f'Frequency spacing: {fft_sample_rate_Hz / fft_count:.2f} Hz; Nyquist: {fft_sample_rate_Hz / 2:.1f} Hz')
fig, axes = plt.subplots(1, 2, figsize=(10, 3), layout='constrained')
axes[0].plot(fft_time_s[:80], fft_voltage_V[:80])
axes[0].set(xlabel='Time (s)', ylabel='Voltage (V)', title='Uniformly sampled signal')
axes[1].stem(fft_frequency_Hz, fft_amplitude_V, basefmt=' ')
axes[1].set(xlabel='Frequency (Hz)', ylabel='Amplitude (V)', title='One-sided amplitude spectrum')
plt.show()
plt.close(fig)
```

### 19.8 Filter a signal using second-order sections


A **low-pass filter** keeps slowly varying signals while reducing rapid changes.
Create a Butterworth filter using a cutoff and sampling rate in hertz. Specifying
`output='sos'` represents it as several small filter stages, called second-order
sections, which help avoid numerical problems.

`sosfiltfilt` filters forwards and backwards. The combined response has no phase
shift and its magnitude is the square of the one-pass response. It uses later
samples, so this is suitable for analysing a completed recording. The beginning
and end can show edge effects, and signals near the cutoff are also attenuated.
Filtering changes noise correlations as well as its size.
[sosfiltfilt](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.sosfiltfilt.html)


```python
filter_rng = np.random.default_rng(238)
filter_rate_Hz = 200.0
filter_time_s = np.arange(1000) / filter_rate_Hz
filter_clean_V = np.sin(2 * np.pi * 3 * filter_time_s)
# Add a 45 Hz disturbance and random noise to the desired 3 Hz signal.
filter_measured_V = filter_clean_V + 0.4 * np.sin(2 * np.pi * 45 * filter_time_s) + filter_rng.normal(0, 0.15, 1000)
# Design a fourth-order low-pass filter with a 10 Hz cutoff. Second-order sections store it as
# smaller filter stages for stable numerical calculation.
filter_sos = signal.butter(4, 10.0, btype='lowpass', fs=filter_rate_Hz, output='sos')
# Apply the filter forwards and backwards to remove phase delay; this uses
# later as well as earlier samples.
filter_smoothed_V = signal.sosfiltfilt(filter_sos, filter_measured_V)
# Leave out the ends, where the finite record and filtering boundary treatment affect the result.
filter_interior = slice(100, -100)  # Exclude boundaries when comparing this synthetic example.
for label, values in [('Measured', filter_measured_V), ('Filtered', filter_smoothed_V)]:
    # Compare against the known clean signal using the root mean square difference, in volts.
    rms_error_V = np.sqrt(np.mean((values[filter_interior] - filter_clean_V[filter_interior])**2))
    print(f'{label} interior RMS error: {rms_error_V:.3f} V')
fig, ax = plt.subplots(figsize=(8, 3), layout='constrained')
ax.plot(filter_time_s, filter_measured_V, alpha=0.4, label='Measured')
ax.plot(filter_time_s, filter_clean_V, '--', label='Known clean signal')
ax.plot(filter_time_s, filter_smoothed_V, label='Filtered')
ax.set(xlim=(1, 2), xlabel='Time (s)', ylabel='Voltage (V)', title='Offline low-pass filtering')
ax.legend()
plt.show()
plt.close(fig)
```

### 19.9 Estimate a power spectral density with Welch averaging


Welch's method estimates how signal power is distributed across frequency.
It splits an evenly sampled recording into overlapping segments, applies a
**window** to reduce sharp changes at each segment's ends, and averages their
power spectra. Longer segments distinguish closer frequencies, but leave fewer
segments to average.

With `scaling='density'`, a voltage input gives a power spectral density in
V²/Hz. Summing the density times the frequency-bin width estimates mean-square
fluctuations. Here that estimate should be close to the sample variance.
Windowing, removing each segment's mean and finite sample size mean that the
two values need not agree exactly.
[welch](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.welch.html)


```python
psd_rng = np.random.default_rng(239)
psd_rate_Hz = 200.0
psd_time_s = np.arange(4096) / psd_rate_Hz
psd_voltage_V = 0.8 * np.sin(2 * np.pi * 25 * psd_time_s) + psd_rng.normal(0, 0.3, len(psd_time_s))
# Estimate spectra from overlapping 512-sample sections and average them. A Hann window tapers each
# section; subtracting its mean removes its constant offset.
psd_frequency_Hz, psd_density_V2_Hz = signal.welch(psd_voltage_V, fs=psd_rate_Hz,
    window='hann', nperseg=512, noverlap=256, detrend='constant', scaling='density')
psd_df_Hz = psd_frequency_Hz[1] - psd_frequency_Hz[0]
# Sum density times frequency-bin width: (V²/Hz)*Hz gives V², comparable with signal variance.
psd_integrated_V2 = np.sum(psd_density_V2_Hz) * psd_df_Hz
print(f'Integrated PSD: {psd_integrated_V2:.3f} V²; sample variance: {np.var(psd_voltage_V):.3f} V²')
fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
# Use a logarithmic vertical scale to see the narrow signal peak and
# lower noise background together.
ax.semilogy(psd_frequency_Hz, psd_density_V2_Hz)
ax.set(xlabel='Frequency (Hz)', ylabel='Power spectral density (V²/Hz)', title='Welch estimate of signal plus noise')
plt.show()
plt.close(fig)
```

### 19.10 Locate pulses using peak prominence and width


`find_peaks` selects local maxima. **Prominence** measures how far a peak rises
above its surrounding baseline; it is different from absolute peak height.
`distance` is in samples, so convert a desired separation in seconds using the
sample interval. `peak_widths` returns fractional sample coordinates, allowing
width markers to be drawn between sample positions.

The plotted width is measured at half prominence; it need not equal a physical
full width at half maximum when the baseline slopes or peaks overlap.
[Peak finding](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.find_peaks.html)


```python
peak_rng = np.random.default_rng(240)
peak_time_s = np.linspace(0, 8, 1601)
peak_dt_s = peak_time_s[1] - peak_time_s[0]
# Create two pulses on a slowly rising background, with a small amount of measurement noise.
peak_voltage_V = (0.03 * peak_time_s + np.exp(-0.5 * ((peak_time_s - 2) / 0.15)**2)
                  + 1.4 * np.exp(-0.5 * ((peak_time_s - 5) / 0.25)**2)
                  + peak_rng.normal(0, 0.015, len(peak_time_s)))
# Prominence measures how far a peak stands above its surrounding baseline. distance is in samples:
# convert the chosen one-second minimum separation using the sampling interval.
peak_indices, peak_properties = signal.find_peaks(peak_voltage_V, prominence=0.5,
                                                  distance=int(np.ceil(1.0 / peak_dt_s)))
# Measure widths at half prominence. The returned left and right positions can lie between samples.
peak_width_samples, peak_levels_V, peak_left, peak_right = signal.peak_widths(peak_voltage_V, peak_indices, rel_height=0.5)
# Convert peak indices to recorded times and widths from samples to seconds.
display(pd.DataFrame({'time_s': peak_time_s[peak_indices], 'prominence_V': peak_properties['prominences'],
                      'half_prominence_width_s': peak_width_samples * peak_dt_s}))
fig, ax = plt.subplots(figsize=(8, 3), layout='constrained')
ax.plot(peak_time_s, peak_voltage_V, label='Pulse record')
ax.plot(peak_time_s[peak_indices], peak_voltage_V[peak_indices], 'x', ms=8, label='Selected peaks')
# Convert the fractional left/right sample positions to times before drawing the width markers.
ax.hlines(peak_levels_V, peak_time_s[0] + peak_left * peak_dt_s,
          peak_time_s[0] + peak_right * peak_dt_s, color='tab:red', label='Half-prominence width')
ax.set(xlabel='Time (s)', ylabel='Voltage (V)', title='Peak positions and widths')
ax.legend()
plt.show()
plt.close(fig)
```

### 19.11 Interpolate a two-dimensional field with explicit axis order


`RegularGridInterpolator` takes one coordinate array per data axis. With
`indexing='ij'`, `field[i, j]` belongs to `(x[i], y[j])`, and query points have
shape **(number of points, 2)** in that same order. This example uses a known
linear temperature field, which linear interpolation can reproduce exactly.

An outside query returns `NaN` here because `bounds_error=False` and
`fill_value=np.nan`. Choose extrapolation deliberately; interpolation alone
does not supply uncertainties or discover the governing physics.
[RegularGridInterpolator](https://docs.scipy.org/doc/scipy/reference/generated/scipy.interpolate.RegularGridInterpolator.html)


```python
field_x_m = np.linspace(0, 2, 9)
field_y_m = np.linspace(0, 1, 7)
# With indexing="ij", axis 0 follows x and axis 1 follows y, giving shape (9, 7).
field_X_m, field_Y_m = np.meshgrid(field_x_m, field_y_m, indexing='ij')
field_temperature_K = 290 + 4 * field_X_m - 3 * field_Y_m
# Supply coordinate arrays in that same axis order. Out-of-range positions
# return NaN instead of extrapolation.
field_interpolator = interpolate.RegularGridInterpolator((field_x_m, field_y_m), field_temperature_K,
    method='linear', bounds_error=False, fill_value=np.nan)
# Each row is one (x, y) query in metres; the last lies outside the sampled x range.
field_queries_m = np.array([[0.15, 0.2], [1.1, 0.7], [1.9, 0.9], [2.2, 0.4]])
field_values_K = field_interpolator(field_queries_m)
# Linear interpolation should reproduce this linear field at all three interior query points.
np.testing.assert_allclose(field_values_K[:3], 290 + 4 * field_queries_m[:3, 0] - 3 * field_queries_m[:3, 1])
assert np.isnan(field_values_K[-1])
display(pd.DataFrame({'x_m': field_queries_m[:, 0], 'y_m': field_queries_m[:, 1], 'temperature_K': field_values_K}))
fig, ax = plt.subplots(figsize=(7, 3.5), layout='constrained')
field_image = ax.pcolormesh(field_X_m, field_Y_m, field_temperature_K, shading='auto', cmap='inferno')
# Use the same temperature colour limits for the queried points and the background field.
ax.scatter(*field_queries_m[:3].T, c=field_values_K[:3], cmap='inferno',
           vmin=field_temperature_K.min(), vmax=field_temperature_K.max(), edgecolors='cyan', s=80, label='Queries')
ax.set(xlabel='x (m)', ylabel='y (m)', aspect='equal', title='Temperature field and interpolated queries')
ax.legend()
fig.colorbar(field_image, ax=ax, label='Temperature (K)')
plt.show()
plt.close(fig)
```

### 19.12 Query nearby detector hits with a spatial tree


`cKDTree` organises point coordinates so you can repeatedly find the nearest
points or all points inside a radius. This avoids storing all possible distances
at once. Both coordinates here are in millimetres, so straight-line distance
has a clear meaning. For mixed units or very many coordinates per observation,
consider carefully whether this distance and search method suit the problem.
[cKDTree](https://docs.scipy.org/doc/scipy/reference/generated/scipy.spatial.cKDTree.html)


```python
tree_rng = np.random.default_rng(242)
tree_hits_mm = tree_rng.uniform(-5, 5, size=(300, 2))
tree_queries_mm = np.array([[0.0, 0.0], [2.0, -1.0]])
# Organise the recorded positions for repeated nearby-point searches.
hit_tree = spatial.cKDTree(tree_hits_mm)
# Find three nearest hits for each query; returned indices identify rows of tree_hits_mm.
tree_distances_mm, tree_indices = hit_tree.query(tree_queries_mm, k=3)
tree_radius_mm = 1.2
# Instead of fixing the number of hits, collect all hits within the
# chosen radius of the first query.
tree_nearby = hit_tree.query_ball_point(tree_queries_mm[0], r=tree_radius_mm)
# As a check, subtract that query from every hit and calculate ordinary Euclidean distances.
tree_brute_distances = np.linalg.norm(tree_hits_mm - tree_queries_mm[0], axis=1)
np.testing.assert_allclose(tree_distances_mm[0], np.sort(tree_brute_distances)[:3])
print('Three nearest indices per query:')
print(tree_indices)
print('Distances (mm):')
print(np.round(tree_distances_mm, 3))
print(f'Hits within {tree_radius_mm} mm of the first query:', len(tree_nearby))
fig, ax = plt.subplots(figsize=(5, 4), layout='constrained')
ax.scatter(*tree_hits_mm.T, s=10, alpha=0.4, label='All hits')
# Use the returned row indices to highlight the selected hits in their original coordinates.
ax.scatter(*tree_hits_mm[tree_nearby].T, s=25, label='Inside radius')
ax.scatter(*tree_queries_mm[0], marker='x', s=80, color='black', label='Query')
ax.add_patch(plt.Circle(tree_queries_mm[0], tree_radius_mm, fill=False, color='tab:red'))
ax.set(xlabel='x (mm)', ylabel='y (mm)', aspect='equal', title='Radius search')
ax.legend()
plt.show()
plt.close(fig)
```

### 19.13 Bootstrap uncertainty on a summary statistic


**Bootstrapping** estimates uncertainty by drawing repeated samples from the
observed readings, with replacement. A reading can appear several times in a
new sample. Here we calculate the mean flight time for each resample and use
the variation between these means to estimate uncertainty in the population mean.

`batch` limits how many resamples are processed together, and `rng` makes the
random choices repeatable. The example uses SciPy's basic bootstrap interval.
Its 95% confidence level describes the intended coverage of intervals over
repeated experiments; it does not mean that 95% of individual readings lie inside.
Resampling rows this way assumes independent, representative observations.
Correlated measurements need a resampling method that respects those correlations.
[bootstrap](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bootstrap.html)


```python
bootstrap_times_us = np.random.default_rng(243).normal(12.0, 0.8, 80)
# Draw repeated samples of 80 flight times, with replacement, and calculate a mean for each. The
# comma in (bootstrap_times_us,) makes a tuple containing one dataset. batch=300 limits how many
# resamples are processed together; it does not change their total number.
bootstrap_result = stats.bootstrap((bootstrap_times_us,), np.mean, vectorized=True,
    confidence_level=0.95, method='basic', n_resamples=3000, batch=300,
    rng=np.random.default_rng(244))
# Read the interval endpoints; standard_error below describes uncertainty in the
# mean, not the spread of flights.
bootstrap_low_us, bootstrap_high_us = bootstrap_result.confidence_interval
print(f'Mean: {bootstrap_times_us.mean():.3f} µs')
print(f'Estimated standard error: {bootstrap_result.standard_error:.3f} µs')
print(f'95% basic bootstrap interval: [{bootstrap_low_us:.3f}, {bootstrap_high_us:.3f}] µs')
fig, ax = plt.subplots(figsize=(7, 3), layout='constrained')
# This histogram contains resampled means, so its horizontal values
# differ from individual flight times.
ax.hist(bootstrap_result.bootstrap_distribution, bins=35, color='tab:blue', alpha=0.7)
ax.axvline(bootstrap_times_us.mean(), color='black', label='Observed mean')
ax.axvline(bootstrap_low_us, color='tab:red', linestyle='--', label='Basic interval limits')
ax.axvline(bootstrap_high_us, color='tab:red', linestyle='--')
ax.set(xlabel='Resampled mean flight time (µs)', ylabel='Resamples per bin', title='Sampling uncertainty of the mean')
ax.legend()
plt.show()
plt.close(fig)
```

### 19.14 Integrate over a region with scrambled Sobol points


Quasi-Monte Carlo methods place points more evenly across a region than
independent random sampling. `Sobol.random_base2(m)` generates $2^m$ points,
preserving the sequence's balance properties. Map the unit square onto a
rectangle, evaluate the density at each point and multiply its mean by the area.

Here the known density is $\rho(x,y)=1+(x/(1\,\mathrm{m}))^2+2(y/(1\,\mathrm{m}))^2$
in kg/m² on a 2 m by 1 m plate, so the exact mass is 6 kg. An independent
**scramble** randomises the point construction while retaining its even coverage.
Repeating with several scrambles estimates variability between the calculated
integrals. The standard error across scrambles is not a strict error bound or
a measurement uncertainty. Points within one scramble are correlated and should
not be treated as independent random observations.
[Sobol sampling](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.qmc.Sobol.html)


```python
qmc_lower_m, qmc_upper_m = np.array([0.0, 0.0]), np.array([2.0, 1.0])
# Multiply the side lengths to find the physical area of the integration region.
qmc_area_m2 = np.prod(qmc_upper_m - qmc_lower_m)
qmc_mass_estimates_kg = []
# Repeat the calculation with eight independent scramblings to estimate numerical variability.
for scramble_seed in range(8):
    sobol_engine = stats.qmc.Sobol(d=2, scramble=True, rng=np.random.default_rng(250 + scramble_seed))
    # m=10 requests 2**10 = 1024 points covering the unit square.
    qmc_unit_points = sobol_engine.random_base2(m=10)
    # Convert unit-square coordinates to the actual rectangular region in metres.
    qmc_points_m = stats.qmc.scale(qmc_unit_points, qmc_lower_m, qmc_upper_m)
    qmc_density_kg_m2 = 1 + qmc_points_m[:, 0]**2 + 2 * qmc_points_m[:, 1]**2
    # Area times average surface mass density estimates total mass: m² times kg/m² gives kg.
    qmc_mass_estimates_kg.append(qmc_area_m2 * qmc_density_kg_m2.mean())
qmc_mass_estimates_kg = np.array(qmc_mass_estimates_kg)
# Estimate uncertainty in the average of the eight mass estimates from their spread.
qmc_standard_error_kg = qmc_mass_estimates_kg.std(ddof=1) / np.sqrt(len(qmc_mass_estimates_kg))
print(f'Mass estimate: {qmc_mass_estimates_kg.mean():.8f} kg; exact: 6 kg')
print(f'Standard error across 8 scrambles: {qmc_standard_error_kg:.2e} kg')
print('Points per scramble:', len(qmc_points_m))
```

### 19.15 Normalise very small weights without numerical underflow


In statistical physics, weights proportional to $e^{-E/(k_BT)}$ can become too
small for the chosen number type to represent and be stored as zero. This is
called **underflow**. Work with logarithms and use `logsumexp` to calculate the
logarithm of the normalising sum. Subtract it before exponentiating to obtain
probabilities. Only energy differences affect the normalised result.

This approach also helps when combining small likelihood values. It improves
the numerical calculation, but the physical or statistical assumptions still
need to be checked.
[logsumexp](https://docs.scipy.org/doc/scipy/reference/generated/scipy.special.logsumexp.html)


```python
energy_over_kBT = np.array([1000.0, 1001.0, 1003.0, 1007.0])
# E/(k_B*T) is dimensionless; the log of a Boltzmann weight is its negative.
boltzmann_log_weights = -energy_over_kBT
# Calculate log(sum(exp(log_weights))) without first forming the tiny
# weights, which would round to zero.
boltzmann_log_partition = special.logsumexp(boltzmann_log_weights)
# Subtract the log of the total weight before exponentiating to obtain normalised probabilities.
boltzmann_probabilities = np.exp(boltzmann_log_weights - boltzmann_log_partition)
print('Direct unnormalised weights:', np.exp(boltzmann_log_weights))
print('Log of the partition sum:', boltzmann_log_partition)
display(pd.DataFrame({'E_over_kBT': energy_over_kBT, 'probability': boltzmann_probabilities}))
np.testing.assert_allclose(boltzmann_probabilities.sum(), 1.0, atol=1e-12)
# A common shift of all log weights must leave relative probabilities unchanged; check that here.
boltzmann_shifted = boltzmann_log_weights + 1000
np.testing.assert_allclose(np.exp(boltzmann_shifted - special.logsumexp(boltzmann_shifted)), boltzmann_probabilities)
```

### 20.1 Create an image and check coordinates and colour order


A colour image has one value per pixel for each of three colour channels.
OpenCV normally stores these as **BGR**: blue, green, red, with array shape
`(height, width, 3)`. NumPy selects pixels as `[row, column]`, or `[y, x]`, while
OpenCV drawing commands take positions as `(x, y)`.

`uint8` stores unsigned 8-bit integers, allowing values from 0 to 255. Convert
to floating point before calculations that could exceed that range.


```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Image axes are (row, column, colour channel); uint8 stores intensities from 0 to 255.
detector_bgr = np.zeros((180, 220, 3), dtype=np.uint8)
# Drawing commands take positions as (x, y), unlike array access [row, column] = [y, x]. OpenCV
# colours are in blue, green, red order; thickness=-1 fills the shape.
cv2.circle(detector_bgr, (55, 60), 20, (30, 180, 240), thickness=-1)
cv2.rectangle(detector_bgr, (115, 30), (175, 80), (200, 220, 80), thickness=-1)
cv2.ellipse(detector_bgr, (135, 130), (28, 14), 20, 0, 360, (180, 80, 220), thickness=-1)
# Convert to red, green, blue order for Matplotlib display, and make a separate intensity image.
detector_rgb = cv2.cvtColor(detector_bgr, cv2.COLOR_BGR2RGB)
detector_gray = cv2.cvtColor(detector_bgr, cv2.COLOR_BGR2GRAY)
print('OpenCV:', cv2.__version__)
print('Colour image:', detector_bgr.shape, detector_bgr.dtype)
print('Pixel at x=55, y=60, as BGR:', detector_bgr[60, 55])
fig, ax = plt.subplots(figsize=(6, 4), layout='constrained')
ax.imshow(detector_rgb)
ax.set(xlabel='Column x (pixels)', ylabel='Row y (pixels)', title='Three synthetic objects')
plt.show()
```

### 20.2 Change colour representation and read image bytes


Matplotlib expects colour values in **RGB** order, so convert an OpenCV BGR
image before displaying it. A grayscale image combines the colour information
into one intensity per pixel; this differs from displaying one colour channel.

`imencode` converts an image to bytes in a file format such as PNG; `imdecode`
reads those bytes back into an array. For an actual image file, use
`cv2.imread(path)` and check that the result is not `None`, which indicates a
failed read.


```python
blue_channel, green_channel, red_channel = cv2.split(detector_bgr)
# Encode an image into PNG bytes in memory, as if saving a file without writing it to disk.
success, png_bytes = cv2.imencode('.png', detector_bgr)
if not success:
    raise RuntimeError('PNG encoding failed')
# Decode those bytes back into pixel values, preserving OpenCV's BGR colour order.
decoded_bgr = cv2.imdecode(png_bytes, cv2.IMREAD_COLOR)
if decoded_bgr is None:
    raise RuntimeError('PNG decoding failed')
print('PNG bytes:', png_bytes.nbytes)
# PNG is lossless: check that encoding and decoding reproduce every pixel exactly.
print('Lossless round trip:', np.array_equal(decoded_bgr, detector_bgr))
fig, axes = plt.subplots(1, 3, figsize=(10, 3), layout='constrained')
# Pair each panel with its image and label; keep grayscale displays on the same intensity scale.
for ax, image, title in zip(axes, [detector_rgb, red_channel, detector_gray],
                            ['RGB display', 'Red-channel intensity', 'Grayscale intensity']):
    ax.imshow(image, cmap='gray', vmin=0, vmax=255)
    ax.set_title(title)
    ax.set_axis_off()
plt.show()
```

### 20.4 Smooth an image and select bright pixels


Gaussian smoothing averages nearby intensities to reduce small-scale fluctuations.
A **threshold** then separates bright objects from darker background: this step
is called **segmentation**. Otsu's method chooses one threshold from the image's
intensity distribution; it does not train a classifier.

The output is a **binary mask**, containing 255 for selected pixels and 0 for
background. Inspect whether the selection matches the objects you intended to
measure, particularly if the illumination varies across the image.


```python
image_rng = np.random.default_rng(181)
# Add noise in floating point and clip before converting back to the stored pixel type.
noisy_gray = np.clip(detector_gray.astype(float) + image_rng.normal(0, 8, detector_gray.shape),
                     0, 255).astype(np.uint8)
# Average neighbouring intensities with Gaussian weights over a 5 by 5 pixel neighbourhood.
smoothed_gray = cv2.GaussianBlur(noisy_gray, (5, 5), sigmaX=1.0)
# Otsu's method chooses a threshold from the intensity distribution. Return both that threshold and
# a mask with values 0 (background) or 255 (selected).
otsu_value, binary_mask = cv2.threshold(smoothed_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print('Selected threshold:', otsu_value, '; mask values:', np.unique(binary_mask))
fig, axes = plt.subplots(1, 3, figsize=(10, 3), layout='constrained')
for ax, image, title in zip(axes, [noisy_gray, smoothed_gray, binary_mask],
                            ['Noisy intensity', 'Gaussian smoothing', 'Binary segmentation']):
    ax.imshow(image, cmap='gray', vmin=0, vmax=255)
    ax.set_title(title)
    ax.set_axis_off()
plt.show()
```

### 20.5 Remove isolated pixels and fill small gaps in a mask


**Morphology** changes shapes in a binary mask using a small neighbourhood of
pixels, called a kernel. **Erosion** shrinks selected regions; **dilation** expands
them. An **opening** applies erosion then dilation to remove small isolated
regions. A **closing** applies dilation then erosion to fill small gaps.

Choose the neighbourhood size in relation to the objects and the pixel scale.
An oversized kernel can remove real objects or merge nearby ones, so inspect
the result before calculating areas.


```python
speckled_mask = binary_mask.copy()
speckle_rows = image_rng.integers(0, speckled_mask.shape[0], 160)
speckle_columns = image_rng.integers(0, speckled_mask.shape[1], 160)
# The row and column arrays specify paired pixel positions where we add isolated bright pixels.
speckled_mask[speckle_rows, speckle_columns] = 255
# Define the small neighbourhood used to shrink and expand selected regions.
morphology_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
# Opening erodes then dilates: it can remove bright features smaller than the neighbourhood.
opened_mask = cv2.morphologyEx(speckled_mask, cv2.MORPH_OPEN, morphology_kernel)
# Closing dilates then erodes: it can fill small dark gaps. Both operations can
# also change object boundaries.
cleaned_mask = cv2.morphologyEx(opened_mask, cv2.MORPH_CLOSE, morphology_kernel)
fig, axes = plt.subplots(1, 2, figsize=(8, 3), layout='constrained')
for ax, image, title in zip(axes, [speckled_mask, cleaned_mask], ['Speckles added', 'After opening and closing']):
    ax.imshow(image, cmap='gray', vmin=0, vmax=255)
    ax.set_title(title)
    ax.set_axis_off()
plt.show()
```

### 20.6 Count connected objects and measure their size and centre


**Connected components** groups touching selected pixels into objects and assigns
each object an integer label. Label 0 is reserved for background. The area is
the number of selected pixels, and the **centroid** is their mean `(x, y)` position.

For physical area, multiply the pixel count by the **square** of the pixel width
for square pixels. The result depends on the threshold and image resolution,
so it is an estimate of the object's area rather than an exact boundary measurement.


```python
# Label groups of connected selected pixels; connectivity=8 includes diagonal neighbours.
component_count, component_labels, component_stats, component_centroids = cv2.connectedComponentsWithStats(
    cleaned_mask, connectivity=8)
pixel_pitch_um = 10.0
component_rows = []
# Label 0 is the background, so begin with object 1.
for object_id in range(1, component_count):
    area_pixels = component_stats[object_id, cv2.CC_STAT_AREA]
    # Ignore objects below this chosen area threshold, such as small remaining speckles.
    if area_pixels < 50:
        continue
    # The centroid averages pixel positions within the object; it is returned in (x, y) order.
    centre_x, centre_y = component_centroids[object_id]
    # Convert pixel count to physical area using the square of the pixel pitch. Use the object's
    # label mask to average its intensity in the original grayscale image.
    component_rows.append({'object_id': object_id, 'area_pixels': area_pixels,
                           'area_um2': area_pixels * pixel_pitch_um**2,
                           'centroid_x_pixels': centre_x, 'centroid_y_pixels': centre_y,
                           'mean_intensity': detector_gray[component_labels == object_id].mean()})
object_measurements = pd.DataFrame(component_rows)
display(object_measurements.round(2))
fig, ax = plt.subplots(figsize=(6, 4), layout='constrained')
ax.imshow(detector_rgb)
# Place each object ID at its measured centre to connect the image with the results table.
for row in component_rows:
    ax.text(row['centroid_x_pixels'], row['centroid_y_pixels'], str(row['object_id']),
             color='white', ha='center', bbox={'facecolor': 'black', 'alpha': 0.7})
ax.set(title='Measured connected objects', xlabel='x (pixels)', ylabel='y (pixels)')
plt.show()
```

### 20.7 Compare edge detection, contours and pixel areas


Canny finds edges where image intensity changes sharply. A **contour** traces
the boundary of a selected region in a binary image. `contourArea` calculates
the area of the polygon enclosed by that boundary. This can differ from counting
selected pixels, so state which definition you used when reporting an area.


```python
image_edges = cv2.Canny(smoothed_gray, threshold1=40, threshold2=100)
# Trace only external boundaries; the second returned item is contour hierarchy, unused here.
image_contours, _ = cv2.findContours(cleaned_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contour_overlay = detector_bgr.copy()
# Draw every contour (-1 selects all) on a copy so the original image is preserved.
cv2.drawContours(contour_overlay, image_contours, -1, (255, 255, 255), thickness=2)
# Compute geometric areas enclosed by the contours; these can differ from counts of selected pixels.
contour_areas = sorted(cv2.contourArea(contour) for contour in image_contours)
print('Contour areas (pixel²):', contour_areas)
fig, axes = plt.subplots(1, 2, figsize=(8, 3), layout='constrained')
axes[0].imshow(image_edges, cmap='gray')
axes[0].set_title('Canny edges')
axes[1].imshow(cv2.cvtColor(contour_overlay, cv2.COLOR_BGR2RGB))
axes[1].set_title('External contours')
for ax in axes:
    ax.set_axis_off()
plt.show()
```

### 20.8 Rotate an image and choose how to calculate new pixels


`getRotationMatrix2D` describes a rotation about a chosen centre, with an optional
change in scale. `warpAffine` uses this transformation to calculate pixels on
the output grid. Content can be cut off if the output is too small.

Interpolation estimates values between input pixels, and border settings decide
what to do beyond the original image. Use nearest-neighbour interpolation for
a mask of category labels so interpolation does not create intermediate labels.


```python
image_height, image_width = detector_gray.shape
# Build a rotation of 20 degrees about the image centre with scale factor 1 (no enlargement).
rotation_matrix = cv2.getRotationMatrix2D((image_width / 2, image_height / 2), 20, 1.0)
# Map into an image of the original size; linear interpolation blends neighbouring colour values.
# Positions outside the source image get a constant border value (zero by default).
rotated_rgb = cv2.warpAffine(detector_rgb, rotation_matrix, (image_width, image_height),
                            flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT)
# Use the nearest source pixel for a mask, preserving discrete 0/255
# values instead of blending them.
rotated_mask = cv2.warpAffine(cleaned_mask, rotation_matrix, (image_width, image_height),
                             flags=cv2.INTER_NEAREST, borderMode=cv2.BORDER_CONSTANT)
print('Affine matrix:', rotation_matrix)
print('Mask still contains discrete values:', np.unique(rotated_mask))
fig, axes = plt.subplots(1, 2, figsize=(8, 3), layout='constrained')
axes[0].imshow(rotated_rgb)
axes[1].imshow(rotated_mask, cmap='gray')
axes[0].set_title('Interpolated image')
axes[1].set_title('Nearest-neighbour mask')
for ax in axes:
    ax.set_axis_off()
plt.show()
```

### 20.9 Find a supplied image patch using template matching


**Template matching** compares a known image patch with each possible location
in a larger image. It is useful when the expected shape, appearance and size
are known. The result gives a score for each possible **top-left** patch position,
so its array is smaller than the original image.

Here the patch is copied from the image itself. This demonstrates how to locate
the best match; it does not show how reliably the method would find an object
under different viewing or lighting conditions. No detector model is trained.


```python
# Copy a known patch to search for; this example takes it from the image itself.
image_template = detector_gray[35:85, 30:80].copy()
# Slide the patch over candidate positions and measure similarity at
# each possible top-left location.
match_scores = cv2.matchTemplate(detector_gray, image_template, cv2.TM_CCOEFF_NORMED)
# minMaxLoc returns minimum, maximum and their locations; this method uses the maximum score.
_, best_score, _, best_location = cv2.minMaxLoc(match_scores)
matched_rgb = detector_rgb.copy()
template_height, template_width = image_template.shape
top_left_x, top_left_y = best_location
# Convert the best top-left position and patch dimensions into opposite corners for the outline.
cv2.rectangle(matched_rgb, best_location,
               (top_left_x + template_width, top_left_y + template_height), (255, 255, 255), 2)
print('Top-left location:', best_location, '; normalised score:', round(best_score, 5))
fig, axes = plt.subplots(1, 2, figsize=(9, 3), layout='constrained')
axes[0].imshow(matched_rgb)
axes[0].set_title('Best template location')
# The map shows correlation scores, not probabilities that an object is present.
score_image = axes[1].imshow(match_scores, cmap='viridis', vmin=-1, vmax=1)
axes[1].set(title='Similarity map', xlabel='Candidate x', ylabel='Candidate y')
fig.colorbar(score_image, ax=axes[1], label='Normalised correlation')
plt.show()
```

### 22.1 Prepare synthetic outcomes and supplied prediction scores

Here category 1 means a signal-like event and is called the **positive
class**; category 0 means background. A supplied formula assigns each event a
score. We will separately check whether scores rank events well, how a cutoff
changes the decisions, and whether scores can be interpreted as probabilities.
The true outcomes are generated using a different, known probability rule.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.model_selection import GroupShuffleSplit, StratifiedGroupKFold, TimeSeriesSplit
from sklearn.preprocessing import FunctionTransformer, normalize
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, classification_report,
                              ConfusionMatrixDisplay, RocCurveDisplay, PrecisionRecallDisplay,
                              roc_auc_score, average_precision_score, brier_score_loss, log_loss,
                              pairwise_distances_chunked, mean_absolute_error, root_mean_squared_error)
from sklearn.calibration import CalibrationDisplay

evaluation_rng = np.random.default_rng(200)
# Create 600 artificial events with two measured inputs per event.
evaluation_features = evaluation_rng.normal(size=(600, 2))
# A sigmoid, 1/(1 + exp(-z)), maps the supplied formula to values between 0 and 1. Use one formula
# to generate labels and a different fixed formula to supply predictions.
underlying_probability = 1 / (1 + np.exp(-(1.2 * evaluation_features[:, 0] - 0.6 * evaluation_features[:, 1] - 1)))
supplied_scores = 1 / (1 + np.exp(-(1.7 * evaluation_features[:, 0] - 0.3 * evaluation_features[:, 1] - 0.7)))
# Compare a uniform random number with each event's true probability to draw a 0/1 outcome.
evaluation_labels = (evaluation_rng.random(600) < underlying_probability).astype(int)
# Assign five consecutive events to each run; these IDs let us keep runs intact when splitting.
acquisition_runs = np.repeat(np.arange(120), 5)
print('Events:', len(evaluation_labels), '; positive-class fraction:', evaluation_labels.mean())
print('Score range:', supplied_scores.min(), supplied_scores.max())
```

### 22.2 Keep acquisition runs together when splitting data


Readings from the same acquisition run can share experimental conditions.
Keep a run together when splitting the data, so closely related observations do
not appear in both training and test groups. `GroupShuffleSplit` does this using
run IDs; its test fraction counts groups. `StratifiedGroupKFold` also tries to
keep similar category proportions in several splits, while keeping runs intact.
We inspect the row indices of these splits without fitting a model.


```python
group_splitter = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42)
# split supplies pairs of row-index arrays; next retrieves the single split requested here.
# GroupShuffleSplit applies test_size to runs, keeping all events from a run together.
reference_indices, heldout_indices = next(group_splitter.split(
    evaluation_features, evaluation_labels, groups=acquisition_runs))
# Convert run IDs to sets and intersect them with &: an empty set means no run crosses the split.
shared_runs = set(acquisition_runs[reference_indices]) & set(acquisition_runs[heldout_indices])
print('Samples on each side:', len(reference_indices), len(heldout_indices))
print('Acquisition runs shared across sides:', shared_runs)

# Make four alternative divisions, keeping runs intact while trying to balance class proportions.
grouped_folds = list(StratifiedGroupKFold(n_splits=4, shuffle=True, random_state=42).split(
    evaluation_features, evaluation_labels, groups=acquisition_runs))
fold_summary = []
# Each division supplies two index arrays; use them to inspect sizes, class balance and shared runs.
for fold_number, (left_indices, right_indices) in enumerate(grouped_folds):
    fold_summary.append({'fold': fold_number, 'heldout_samples': len(right_indices),
                         'positive_fraction': evaluation_labels[right_indices].mean(),
                         'overlapping_runs': len(set(acquisition_runs[left_indices]) & set(acquisition_runs[right_indices]))})
display(pd.DataFrame(fold_summary))
```

### 22.3 Split in time order and leave a gap between groups


`TimeSeriesSplit` puts earlier observations in one group and later ones in
the test group, expanding the earlier group for each split. `gap` leaves a chosen
number of rows between them. The times here are equally spaced, so two rows also
represent a known duration. With irregular sampling, choose boundaries from the
timestamps and the physical time separation you need.


```python
ordered_times = pd.date_range('2026-01-01', periods=24, freq='1min')
# Keep earlier rows on one side and four later rows on the other, leaving two unused rows between.
temporal_splits = list(TimeSeriesSplit(n_splits=3, test_size=4, gap=2).split(ordered_times))
temporal_rows = []
# Inspect each division in time order; [-1] takes the final earlier index
# and [0] the first later index.
for window, (earlier_indices, later_indices) in enumerate(temporal_splits):
    # Calculate the separation from the indices, then report the corresponding physical timestamps.
    temporal_rows.append({'window': window, 'earlier_end': ordered_times[earlier_indices[-1]],
                          'later_start': ordered_times[later_indices[0]],
                          'excluded_rows': later_indices[0] - earlier_indices[-1] - 1,
                          'later_count': len(later_indices)})
display(pd.DataFrame(temporal_rows))
```

### 22.4 Apply fixed processing steps in a Pipeline


A **Pipeline** applies processing steps in a chosen order. Here each step
uses a supplied calculation, with no parameters to learn from the data. A
`FunctionTransformer` lets that calculation be used as a pipeline step.

First, `log1p` calculates log(1 + count) to compress the range of dimensionless
counts. Then each row is divided by its Euclidean length. This retains relative
channel strengths but removes the overall signal size, so use it only if that
size is irrelevant to your analysis. The example calls `transform` without `fit`.
[FunctionTransformer](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.FunctionTransformer.html)


```python
positive_channel_counts = np.array([[1., 4., 9.], [3., 2., 5.], [10., 20., 30.]])
# List the processing steps in the order they should run; the names identify each step.
stateless_pipeline = Pipeline([
    # log1p computes log(1 + counts), compressing large values while allowing zeros.
    ('compress_counts', FunctionTransformer(np.log1p, inverse_func=np.expm1)),
    # Normalise each row separately to Euclidean length one; this removes
    # overall magnitude information.
    ('unit_length', FunctionTransformer(normalize, kw_args={'norm': 'l2', 'axis': 1})),
])
# These operations use fixed rules and need no fitted parameters;
# transform applies them in sequence.
transformed_counts = stateless_pipeline.transform(positive_channel_counts)
print('Input counts:', positive_channel_counts)
print('Transformed rows:', transformed_counts)
# Check the row lengths after processing; each should be one.
print('Row lengths:', np.linalg.norm(transformed_counts, axis=1))
```

### 22.5 Find nearby reference points without storing a full distance matrix

Find the nearest reference point for each query point, calculating distances
in smaller blocks with `pairwise_distances_chunked`. The helper function keeps
only the nearest point's index and distance from each block. Both coordinates
use millimetres, so the distances can be compared directly. `working_memory`
sets an approximate memory budget in MiB for the distance blocks, not a limit
on the whole Python session.


```python
reference_positions_mm = np.array([[0., 0.], [1., 0.], [0., 1.], [1., 1.]])
query_positions_mm = evaluation_rng.uniform(-0.2, 1.2, (31, 2))

# The library calls this function for each block of distances. Rows are query points, columns are
# reference points, and start is the first original query-row index.
def nearest_reference(distance_chunk, start):
    # Find the column containing the shortest distance in each query row.
    reference_index = distance_chunk.argmin(axis=1)
    # Pair each row index with its selected column index to retrieve one distance per query.
    distance_mm = distance_chunk[np.arange(len(distance_chunk)), reference_index]
    query_index = np.arange(start, start + len(distance_chunk))
    return np.column_stack([query_index, reference_index, distance_mm])

# Process small blocks under a memory budget, keeping only each query's nearest result. vstack joins
# the returned rows; we never retain the complete distance matrix.
nearest_rows = np.vstack(list(pairwise_distances_chunked(
    query_positions_mm, reference_positions_mm, reduce_func=nearest_reference, working_memory=0.001)))
nearest_table = pd.DataFrame(nearest_rows, columns=['query_index', 'reference_index', 'distance_mm'])
# Combining indices with distances made all columns floating point;
# restore integer types for row IDs.
nearest_table[['query_index', 'reference_index']] = nearest_table[['query_index', 'reference_index']].astype(int)
display(nearest_table.head())
```

### 22.6 Check category predictions and count different mistakes


A **confusion matrix** counts correct decisions and the two kinds of mistake:
background selected as signal, and signal missed as background. Dividing each row

- **Accuracy:** the fraction of all decisions that are correct.
- **Precision:** the fraction of selected signal candidates that really are signal.
- **Recall:** the fraction of all true signal events that were selected.
- **Balanced accuracy:** calculate recall for each category and average those values.

These answer different questions, especially when signal events are rare.


```python
# Convert the supplied scores into class IDs using a chosen 0.5 threshold.
supplied_decisions = (supplied_scores >= 0.5).astype(int)
print('Accuracy:', accuracy_score(evaluation_labels, supplied_decisions))
# Ordinary accuracy counts all events equally; balanced accuracy averages the two class recalls.
print('Balanced accuracy:', balanced_accuracy_score(evaluation_labels, supplied_decisions))
print(classification_report(evaluation_labels, supplied_decisions,
                            target_names=['Background-like', 'Signal-like'], zero_division=0))
fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout='constrained')
# Show counts beside row-normalised fractions. Each row refers to one true class, so its diagonal
# entry in the second panel is the fraction correctly identified.
for ax, normalisation, title in zip(axes, [None, 'true'], ['Event counts', 'Fractions within each true class']):
    ConfusionMatrixDisplay.from_predictions(evaluation_labels, supplied_decisions, labels=[0, 1],
                                             normalize=normalisation, colorbar=False, ax=ax)
    ax.set_title(title)
plt.show()
```

### 22.9 Compare missed signal with false alarms when choosing a threshold


Lowering a threshold can accept more signal but also more background.
Raising it can reject background while missing genuine signal. This example
assigns a missed signal five times the cost of a false alarm to illustrate that
choice. A real threshold should reflect the scientific purpose and be chosen
using separate development data or a rule fixed beforehand, before checking the
final test results.


```python
threshold_rows = []
# Try three possible cuts on the same supplied scores.
for threshold in [0.25, 0.5, 0.75]:
    decisions = supplied_scores >= threshold
    # & requires both conditions: selected as signal, but actually labelled background.
    false_positives = np.count_nonzero(decisions & (evaluation_labels == 0))
    # ~ reverses the selection: missed events that were actually labelled signal.
    false_negatives = np.count_nonzero(~decisions & (evaluation_labels == 1))
    # For illustration, assign a missed signal five times the cost of a false alarm. This chosen
    # weighting is part of the analysis question, not a property learned from the data.
    threshold_rows.append({'threshold': threshold, 'false_positives': false_positives,
                           'false_negatives': false_negatives,
                           'illustrative_cost_per_event': (false_positives + 5 * false_negatives) / len(decisions)})
threshold_comparison = pd.DataFrame(threshold_rows)
display(threshold_comparison)
```

## 23. PyTorch data handling and forward evaluation

PyTorch works with **tensors**, which are arrays with one or more dimensions.
Here we use it to select measurements, collect them into groups and pass them

Run the section in order. It creates its own data and runs on the CPU, so no

each dimension represents. A **batch** is a group of observations processed
together; a **channel** is a separate measurement or image colour component.

| Data | Order of dimensions | Example shape |
| --- | --- | --- |
| Measurement table | `(samples, features)` | `(25, 4)`: 25 events, four readings each |
| Time records | `(batch, time, channels)` | `(3, 8, 4)`: three events, eight times, four channels |

### 23.1 Create tensors and inspect their shape, type and device

A tensor is a multidimensional array. Use floating-point values for continuous
measurements and integer values for class IDs or indices. `shape` describes the
axes; `dtype` describes the stored numbers; `device` says where they are stored.
These examples use `float32` and a CPU. NumPy commonly starts with `float64`.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import Dataset, TensorDataset, DataLoader, Subset
from IPython.display import display

torch_rng = torch.Generator().manual_seed(42)
# A tensor is an array used by PyTorch. Here rows are two events and columns are four channels.
# Store measured values as floating-point numbers and class IDs as integers.
readings = torch.tensor([[0.2, 0.4, 0.6, 0.8],
                         [0.3, 0.5, 0.7, 0.9]], dtype=torch.float32)
class_ids = torch.tensor([0, 1], dtype=torch.int64)
print('PyTorch:', torch.__version__)
# Inspect shape, numerical type and device (CPU or GPU) before passing data into a model.
print('readings:', readings.shape, readings.dtype, readings.device)
print('class IDs:', class_ids.shape, class_ids.dtype)
print('Number of axes:', readings.ndim, '; stored values:', readings.numel())
```

### 23.2 Convert between NumPy and PyTorch without accidental changes

`torch.from_numpy` shares the NumPy array's memory. Editing either object changes
the shared values. `clone()` makes an independent tensor. `torch.tensor(array)`
also copies its input. Before sending model results to NumPy, use
`result.detach().cpu().numpy()`; add `.copy()` if NumPy needs independent storage.


```python
numpy_readings = np.array([1.0, 2.0, 3.0], dtype=np.float32)
# from_numpy shares the array's stored values, so editing either object changes both.
shared_readings = torch.from_numpy(numpy_readings)
# clone creates separate storage, useful when keeping an unmodified reference measurement.
independent_readings = shared_readings.clone()
shared_readings[0] = 99.0
print('Shared NumPy data:', numpy_readings)
print('Independent tensor:', independent_readings)
# detach separates any gradient history; cpu puts values in CPU memory for NumPy. The final copy
# prevents edits to the NumPy result from changing this tensor.
numpy_copy = independent_readings.detach().cpu().numpy().copy()
numpy_copy[1] = -1.0
print('Tensor after editing NumPy copy:', independent_readings)
```

### 23.3 Select rows, columns and values with masks

Integer indexing removes an axis; a slice keeps it. Thus `x[0]` is one feature
vector, while `x[0:1]` is a batch containing one vector. A Boolean mask selects
matching entries. Clone first when an edit should leave the source unchanged.


```python
# An integer index removes the event axis, whereas the slice 0:1 keeps a one-event batch.
print('One row:', readings[0], readings[0].shape)
print('One-row batch:', readings[0:1].shape)
print('Second channel across samples:', readings[:, 1])
print('Selected channels:', readings[:, [0, 3]])
# Use channel 0 to decide which events to keep, then retain all channels for those events.
selected_rows = readings[:, 0] > 0.25
print('Selected samples:', readings[selected_rows])

# Work on a copy so a calibration or selection experiment does not overwrite the original readings.
edited_readings = readings.clone()
edited_readings[:, 0] *= 2.0           # Modify one channel for every sample.
# Cap every value above 0.8 at 0.8, wherever it occurs in the array.
edited_readings[edited_readings > 0.8] = 0.8
print('Edited copy:', edited_readings)
print('Original data:', readings)
```

### 23.4 Add, remove, reorder and combine dimensions

`unsqueeze` adds a dimension of length one. `squeeze(dim=...)` removes a
chosen dimension if its length is one; specify it so a one-event batch does not
accidentally lose its batch dimension. `permute` changes the order of dimensions,
for example swapping time and channel positions.

`reshape` regroups the values in their current order; it does not swap the
physical meanings of axes. It can copy values when needed. `view` is more
restrictive because it must use the existing arrangement in memory. `stack`
adds a dimension to hold several tensors; `cat` joins them along an existing one.


```python
# unsqueeze(0) adds an event axis of length one; squeeze(dim=0) removes only that axis.
one_batch = readings[0].unsqueeze(0)
print('Add/remove batch axis:', one_batch.shape, one_batch.squeeze(dim=0).shape)

# Use axis order (event, time, channel): two events, three time samples and four channels.
sequence_demo = torch.arange(24, dtype=torch.float32).reshape(2, 3, 4)
channels_first = sequence_demo.permute(0, 2, 1)  # (batch, channels, time)
# reshape regroups values in their existing order; it does not exchange time
# and channel axes like permute.
regrouped = sequence_demo.reshape(2, 4, 3)      # Same shape, different mapping!
print('Permuted first channel:', channels_first[0, 0])
print('Reshaped first row:', regrouped[0, 0])
print('Flatten all but batch:', sequence_demo.flatten(start_dim=1).shape)
# stack adds a new axis, while cat joins entries along an existing axis.
print('Stack two batches:', torch.stack([readings, readings], dim=0).shape)
print('Join samples:', torch.cat([readings, readings], dim=0).shape)

# Average over time separately for each event and channel. Keep the time axis at length one so
# subtracting this mean repeats the correct channel baseline at every time sample.
channel_mean = sequence_demo.mean(dim=1, keepdim=True)  # (2, 1, 4)
centred_sequence = sequence_demo - channel_mean         # Broadcast over time.
print('Mean shape:', channel_mean.shape, '; centred shape:', centred_sequence.shape)
print('Mean after centring:', centred_sequence.mean(dim=1))
```

### 23.5 Build a small labelled TensorDataset

Each synthetic event has four normalised detector-channel readings. The reference
label identifies which pair of **clean** channels has the greater mean signal.
The model will receive noisy measurements. These labels are an illustrative
rule, not a real particle-identification dataset.

`TensorDataset` pairs corresponding rows. It stores samples; it does not fit a
model. There are 25 events so a batch size of 6 will leave a final batch of one.


```python
# Create 25 four-channel events, then add measurement noise while retaining the clean reference.
clean_signals = torch.rand((25, 4), generator=torch_rng)
observed_signals = clean_signals + 0.08 * torch.randn((25, 4), generator=torch_rng)
# For each event, compare the mean of channels 0/1 with the mean of channels 2/3. stack(..., dim=1)
# places these two scores side by side, giving shape (25, 2).
reference_scores = torch.stack([
    clean_signals[:, :2].mean(dim=1),
    clean_signals[:, 2:].mean(dim=1),
], dim=1)
# argmax returns the position of the larger score: class 0 or class 1 for each event.
event_labels = reference_scores.argmax(dim=1)
# Pair noisy readings with their reference labels; item i returns entries
# from the same row in both tensors.
event_dataset = TensorDataset(observed_signals, event_labels)

sample_features, sample_label = event_dataset[0]
print('Number of samples:', len(event_dataset))
print('Sample 0:', sample_features, '; label:', sample_label.item())
print('Sample shapes:', sample_features.shape, sample_label.shape)
```

### 23.6 Write a Dataset with named fields and safe sample access

A custom `Dataset` tells PyTorch how to count and retrieve observations.
`__len__` supplies the number of samples; `__getitem__` retrieves the sample at a
given index. Returning a dictionary gives names to the measurements, label and
event ID. Here each retrieved measurement is copied with `clone()` so editing it
does not change the stored dataset. Larger datasets can read one sample from a
file when it is requested.


```python
# Define how one recorded event is stored and retrieved using PyTorch's Dataset interface.
class DetectorDataset(Dataset):
    # This runs when a dataset is created; self refers to the dataset being built.
    def __init__(self, features, labels):
        if len(features) != len(labels):
            raise ValueError('Features and labels must have matching sample counts.')
        # Keep independent copies so later edits to the supplied arrays do
        # not change the stored dataset.
        self.features = features.clone()
        self.labels = labels.clone()

    # len(dataset) calls this method to find how many events are available.
    def __len__(self):
        return len(self.features)

    # dataset[index] calls this method. Return named fields for one event, keeping its ID. Clone the
    # retrieved tensors so editing a returned sample does not alter stored measurements.
    def __getitem__(self, index):
        return {
            'features': self.features[index].clone(),
            'label': self.labels[index].clone(),
            'event_id': index,
        }

named_events = DetectorDataset(observed_signals, event_labels)
event_zero = named_events[0]
# Deliberately change a returned reading, then inspect the stored sample to confirm it is unchanged.
event_zero['features'][0] = -999.0
print('Edited retrieved sample:', event_zero)
print('Stored sample remains:', named_events[0])

# Choose four events by their original indices. Subset item 1 refers to original event 3.
selected_events = Subset(named_events, [0, 3, 7, 12])
print('Subset length:', len(selected_events))
print('Subset item 1 has original event ID:', selected_events[1]['event_id'])
```

### 23.7 Access a DataLoader batch and iterate over the data

A `DataLoader` collects samples into **batches**, adding a dimension for
the samples in each group. It keeps dictionary entries with the same name together.
`iter(loader)` starts an iterator and `next(...)` requests its next batch; a new
`for` loop starts again from the loader.

Use `shuffle=False` to keep event order and `drop_last=False` to include the
smaller final batch. `num_workers=0` loads these small examples within the current
Python process. Changing the loading order does not train a model.


```python
# Collect up to six events per batch, preserving their order. Keep the final one-event batch;
# num_workers=0 loads data in this notebook process, without starting worker processes.
event_loader = DataLoader(named_events, batch_size=6, shuffle=False,
                          drop_last=False, num_workers=0)
# iter creates a fresh pass through the loader and next retrieves its first batch. Matching
# dictionary fields are combined: features have shape (6, 4), labels shape (6,).
first_batch = next(iter(event_loader))
print('Batch fields:', list(first_batch))
print('Features:', first_batch['features'].shape)
print('Labels:', first_batch['label'].shape)

batch_sizes = []
# A new loop starts a fresh pass, so the first batch is included again in this full count.
for batch_number, batch in enumerate(event_loader):
    batch_sizes.append(len(batch['label']))
    print('Batch', batch_number, 'shape', tuple(batch['features'].shape),
          'event IDs', batch['event_id'].tolist())
print('Total samples visited:', sum(batch_sizes))

# A TensorDataset returns paired values rather than named fields; unpack them into two variables.
tuple_loader = DataLoader(event_dataset, batch_size=6)
tuple_features, tuple_labels = next(iter(tuple_loader))
# Shuffle the event order reproducibly while keeping each event's fields together.
shuffled_loader = DataLoader(named_events, batch_size=6, shuffle=True,
                             generator=torch.Generator().manual_seed(7))
print('A shuffled batch of IDs:', next(iter(shuffled_loader))['event_id'].tolist())
```

### 23.8 Define and inspect a model with supplied fixed weights

This model uses `nn.Linear(4, 2)` to turn four channel readings into two
scores: the means of two pairs of channels. We supply the coefficients and hold
them fixed. Nothing is learned from the data.

Use `model(inputs)` to calculate outputs. `model.eval()` selects evaluation
behaviour for layers that work differently during training. It does not stop
PyTorch keeping information needed for derivatives; `torch.inference_mode()`
does that during these forward calculations. The fixed linear layer itself
behaves the same in either mode, but this calling pattern also suits later models.
[PyTorch inference mode](https://docs.pytorch.org/docs/stable/generated/torch.autograd.grad_mode.inference_mode.html)


```python
class FixedChannelClassifier(nn.Module):
    def __init__(self):
        # Initialise the PyTorch model machinery before adding its components.
        super().__init__()
        # Map four input channels to two output scores. bias=False omits an added constant offset.
        self.readout = nn.Linear(4, 2, bias=False)
        # Supply a known averaging rule, not an estimate from the examples.
        # Set the weights directly without recording gradient calculations. Each matrix row averages
        # one pair of channels, so the two outputs have a known meaning.
        with torch.no_grad():
            self.readout.weight.copy_(torch.tensor([
                [0.5, 0.5, 0.0, 0.0],
                [0.0, 0.0, 0.5, 0.5],
            ]))
        # Keep these supplied weights fixed; this example only evaluates their rule.
        self.requires_grad_(False)

    # Calling model(features) runs forward; any event rows stay together
    # while the last axis changes 4 -> 2.
    def forward(self, features):
        return self.readout(features)

# Choose evaluation mode. eval changes mode-dependent behaviour; it does
# not calculate predictions itself.
fixed_model = FixedChannelClassifier().eval()
print(fixed_model)
for name, parameter in fixed_model.named_parameters():
    print(name, tuple(parameter.shape), 'requires_grad:', parameter.requires_grad)
    print(parameter)
```

### 23.9 Evaluate one event and a batch without updating the model

Keep a dimension for the number of events even when there is only one:
the input shape is `(1, 4)` and the output is `(1, 2)`. `argmax(dim=1)` picks the
position of the larger score in each row, giving a category ID. `softmax(dim=1)`
turns the row's scores into non-negative values that sum to one. This alone does
not establish that they are reliable probabilities for physical event types.


```python
# Evaluate without recording information for gradient calculations; no parameters are updated.
with torch.inference_mode():
    # The slice 0:1 keeps shape (1, 4), so one event follows the same layout as a larger batch.
    one_event_scores = fixed_model(observed_signals[0:1])
    batch_scores = fixed_model(first_batch['features'])
    # softmax turns each event's two scores into positive values summing to one. This normalisation
    # alone does not establish that they are reliable physical probabilities.
    batch_normalised_scores = torch.softmax(batch_scores, dim=1)
    # Choose the larger score along the class axis (dim=1), returning one class ID per event.
    batch_predictions = batch_scores.argmax(dim=1)

print('One-event output:', one_event_scores, one_event_scores.shape)
print('Batch output:', batch_scores.shape)
print('Predicted class IDs:', batch_predictions)
print('Row sums after softmax:', batch_normalised_scores.sum(dim=1))
print('Output requires gradients:', batch_scores.requires_grad)
```

### 23.10 Evaluate every batch and collect predictions in sample order

Calculate predictions for every batch, keeping them in event order.
To find accuracy, count correctly predicted events and divide by the total event
count. Averaging batch accuracies equally would give the final one-event batch
too much weight. Keep event IDs with results so you can find the original readings.


```python
# Collect answers, reference labels and event IDs together so each
# result can be traced to its input.
prediction_batches, label_batches, id_batches = [], [], []
number_correct = 0
number_evaluated = 0
with torch.inference_mode():
    # shuffle=False preserves sample order, including the smaller final batch.
    for batch in event_loader:
        # Apply the same fixed model to every row of this batch, producing two scores per event.
        scores = fixed_model(batch['features'])
        predicted_labels = scores.argmax(dim=1)
        # Compare class IDs, sum the True matches, then use item() to
        # obtain an ordinary Python number.
        number_correct += (predicted_labels == batch['label']).sum().item()
        # Count actual events in each batch, so the smaller last batch gets
        # the correct weight in accuracy.
        number_evaluated += len(batch['label'])
        # Keep results in CPU memory for the later table.
        prediction_batches.append(predicted_labels.cpu())
        label_batches.append(batch['label'].cpu())
        id_batches.append(batch['event_id'])

# cat joins batch results along the event axis, giving one prediction for each of the 25 events.
all_predictions = torch.cat(prediction_batches)
all_reference_labels = torch.cat(label_batches)
# Use the retained IDs and labels to build an event-by-event record of the evaluation.
evaluation_table = pd.DataFrame({
    'event_id': torch.cat(id_batches).numpy(),
    'reference_label': all_reference_labels.numpy(),
    'predicted_label': all_predictions.numpy(),
})
print(f'Correct: {number_correct}/{number_evaluated} ({number_correct / number_evaluated:.1%})')
display(evaluation_table.head(8))
print('This checks a fixed rule on synthetic noisy signals, not general model performance.')
```

### 23.12 Access time and channel dimensions inside a forward method

The input dimensions are `(batch, time, channels)`. First subtract the same
supplied channel offsets at every time and for every event, using broadcasting.
The linear layer combines the last dimension, leaving the event and time axes
in place. Taking the mean along `dim=1` then gives two scores per event.

`register_buffer` stores the fixed offsets with the model so they move with it
if the computing device changes. They are fixed values, not quantities to train.


```python
class FixedSequenceModel(nn.Module):
    def __init__(self):
        super().__init__()
        # Store fixed channel offsets as part of the model. A buffer moves with the model between
        # devices and is saved with it, but is not a parameter to be learned.
        self.register_buffer('channel_offset', torch.tensor([0.1, 0.0, 0.05, 0.0]))
        self.readout = FixedChannelClassifier()

    def forward(self, values):
        # Subtract the four channel offsets from every time sample of every event by broadcasting.
        corrected = values - self.channel_offset     # (batch, time, 4)
        # The readout acts on the final channel axis, preserving the event and time axes.
        per_time_scores = self.readout(corrected)    # (batch, time, 2)
        return {
            # Keep all events and channels at the first time sample: shape (batch, 4).
            'first_time_channels': corrected[:, 0, :],
            # Keep all times for channel 0: shape (batch, time).
            'channel_zero_trace': corrected[:, :, 0],
            'per_time_scores': per_time_scores,
            # Average over the time axis to obtain one pair of scores per event: shape (batch, 2).
            'event_scores': per_time_scores.mean(dim=1),
        }

# Supply three events, each with eight time samples from four channels.
sequence_inputs = torch.rand((3, 8, 4), generator=torch_rng)
sequence_model = FixedSequenceModel().eval()
with torch.inference_mode():
    sequence_outputs = sequence_model(sequence_inputs)
for name, values in sequence_outputs.items():
    print(name, tuple(values.shape))
```

### 23.13 Reorder image axes, crop, reduce and flatten inside a model

This model summarises images with fixed operations. It receives colour
channels in the last dimension and moves them before the spatial dimensions.
The last two axes then represent image height and width. `mean(dim=(-2, -1))`
averages over both, leaving one mean for each image and channel.

`[..., 1:-1, 1:-1]` removes the outer row and column on each side. Here `...`
means keep all earlier dimensions. `flatten(start_dim=1)` combines the remaining
dimensions into one list of values per image, keeping the batch dimension.
This example calculates image summaries without classifying or training.


```python
class ImageSummary(nn.Module):
    def forward(self, channels_last_images):
        # Change (batch, height, width, channels) to (batch, channels, height, width). permute
        # reorders axes while keeping each pixel with its original channel and image.
        images = channels_last_images.permute(0, 3, 1, 2)
        # ... keeps the batch and channel axes; 1:-1 removes one
        # border pixel from each spatial edge.
        central_crop = images[..., 1:-1, 1:-1]
        return {
            'channels_first': images,
            'central_crop': central_crop,
            # The last two axes are height and width: average both to get
            # one mean per image and channel.
            'channel_means': images.mean(dim=(-2, -1)),
            # Combine channel and spatial axes into one list of values per
            # image, keeping the batch axis.
            'flattened_crop': central_crop.flatten(start_dim=1),
        }

# Generate two images of height 6 and width 8, each with three channels.
image_inputs = torch.rand((2, 6, 8, 3), generator=torch_rng)
image_model = ImageSummary().eval()
with torch.inference_mode():
    image_outputs = image_model(image_inputs)
    # Keep the batch axis even for one image, so the same forward function applies.
    single_image_outputs = image_model(image_inputs[0:1])
for name, values in image_outputs.items():
    print(name, tuple(values.shape))
print('Single-image channel means:', single_image_outputs['channel_means'].shape)

fig, axes = plt.subplots(1, 3, figsize=(9, 3), layout='constrained')
# Select image 0 and display each channel separately on a shared 0 to 1 intensity scale.
for channel, ax in enumerate(axes):
    image = ax.imshow(image_outputs['channels_first'][0, channel].cpu().numpy(),
                      vmin=0, vmax=1, origin='lower', cmap='viridis')
    ax.set(title=f'Image 0, channel {channel}', xlabel='Column index', ylabel='Row index')
fig.colorbar(image, ax=axes, label='Normalised response (dimensionless)', shrink=0.75)
plt.show()
```
